">
Full, Task-Appropriate Evaluation
</div>

<div style="
    font-size:13px;
    line-height:1.55;
    color:#475569;
">
A professional model evaluation should use metrics that match the task instead of relying on one number.

For our IMDb sentiment classifier, we will evaluate performance using
<span style="color:#2563EB;font-weight:700;">Precision</span>,
<span style="color:#DB2777;font-weight:700;">Recall</span>,
<span style="color:#059669;font-weight:700;">F1-score</span>,
and
<span style="color:#D97706;font-weight:700;">AUC-ROC</span>.

Accuracy will still be reported, but it will not be treated as the only measure of model quality.
</div>

</div>

<div style="

Classification Metrics
</div>

<table style="
    width:100%;
    border-collapse:collapse;
    font-size:13px;
">

<tr>
<th style="padding:11px;background:#F8FAFC;border:1px solid #E2E8F0;">Metric</th>
<th style="padding:11px;background:#F8FAFC;border:1px solid #E2E8F0;">What It Tells Us</th>
</tr>

<tr>
<td style="padding:11px;border:1px solid #E2E8F0;"><b>Accuracy</b></td>
<td style="padding:11px;border:1px solid #E2E8F0;">How often the model is correct overall.</td>
</tr>

<tr>
<td style="padding:11px;border:1px solid #E2E8F0;"><b>Precision</b></td>
<td style="padding:11px;border:1px solid #E2E8F0;">Of all reviews predicted as Positive, how many were actually Positive?</td>
</tr>

<tr>
<td style="padding:11px;border:1px solid #E2E8F0;"><b>Recall</b></td>
<td style="padding:11px;border:1px solid #E2E8F0;">Of all truly Positive reviews, how many did the model correctly identify?</td>
</tr>

<tr>
<td style="padding:11px;border:1px solid #E2E8F0;"><b>F1-score</b></td>
<td style="padding:11px;border:1px solid #E2E8F0;">A balance between Precision and Recall.</td>
</tr>

<tr>
<td style="padding:11px;border:1px solid #E2E8F0;"><b>AUC-ROC</b></td>
<td style="padding:11px;border:1px solid #E2E8F0;">How well the model separates Positive and Negative reviews across decision thresholds.</td>

Using several complementary metrics gives a more complete picture of performance and helps reveal weaknesses that overall accuracy may hide.

Why Compare Against a Baseline?
</div>

<div style="
    font-size:13px;
    line-height:1.55;
    color:#475569;
">
A model should not be judged only by its current performance.

To understand whether the final workflow provides real value, we compare it against an earlier
<span style="color:#7C3AED;font-weight:700;">baseline</span>.

The comparison helps us answer a simple question:

<b>Did the model actually improve, and on which metrics?</b>
</div>

</div>

">
Baseline Comparison Note
</div>

<div style="
    font-size:13px;
    line-height:1.55;
    color:#475569;
">
A direct numerical comparison with the Week 6 baseline is not appropriate because the Week 6 model was built for a different dataset and task.

For a fair comparison, models should be evaluated on the
<b>same dataset</b>,
<b>same train/test split</b>,
and
<b>same evaluation metric</b>.

Therefore, the current IMDb model is evaluated using its own task-appropriate metrics, while previous results from different projects are not treated as a valid baseline comparison.

</div>

<div style="

<div style="

Because Day 5 is a separate notebook, the IMDb preprocessing function, TF-IDF vectorizer, trained Logistic Regression model, and test data are not automatically available.

We first recreate the same workflow used in the previous notebook so the evaluation and explainability steps operate on a consistent model.

```python
import re
import html
import nltk

from nltk.corpus import stopwords, wordnet
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

resources = [
    "punkt",
    "punkt_tab",
    "stopwords",
    "wordnet",
    "omw-1.4",
    "averaged_perceptron_tagger_eng"
]

for resource in resources:
    nltk.download(resource, quiet=True)

negation_words = {"not", "no", "nor"}
stop_words = set(stopwords.words("english")) - negation_words
lemmatizer = WordNetLemmatizer()


def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("N"):
        return wordnet.NOUN
    elif tag.startswith("R"):
        return wordnet.ADV
    else:
        return wordnet.NOUN


def preprocess_review(text):
    text = html.unescape(str(text))
    text = re.sub(r"<[^>]+>", " ", text)
    text = re.sub(r"n['’]t\b", " not", text)
    text = text.lower()
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"[^\w\s']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    tokens = word_tokenize(text)

    tokens = [
        token for token in tokens
        if token not in stop_words and len(token) > 1
    ]

    tagged_tokens = pos_tag(tokens)

    lemmas = [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tagged_tokens
    ]

    lemmas = [
        word for word in lemmas
        if not word.startswith("'")
    ]

    return " ".join(lemmas)

print("Preprocessing pipeline ready.")
```

```python
import pandas as pd
from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Load IMDb dataset
data_path = Path("../Day1/IMDB Dataset.csv")

df = pd.read_csv(data_path)
df = df.drop_duplicates().reset_index(drop=True)

# Apply preprocessing
df["processed_review"] = df["review"].apply(preprocess_review)

# Numeric labels
df["label"] = df["sentiment"].map({
    "negative": 0,
    "positive": 1
})

# Keep original reviews for later SHAP/error analysis
train_df, test_df = train_test_split(
    df[["review", "processed_review", "label"]],
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)

X_train = train_df["processed_review"]
X_test = test_df["processed_review"]

y_train = train_df["label"]
y_test = test_df["label"]

# TF-IDF
vectorizer = TfidfVectorizer(max_features=5000)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# Logistic Regression
baseline_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

baseline_model.fit(X_train_tfidf, y_train)

print("IMDb model components are ready.")
print("Training shape:", X_train_tfidf.shape)
print("Testing shape:", X_test_tfidf.shape)
```
