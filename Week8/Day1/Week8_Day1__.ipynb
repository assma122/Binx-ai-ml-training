">
1. Why Does Text Need Preprocessing?
</h2>

<div style="
    width:72px;
    height:4px;
    margin:14px 0 20px 0;
    border-radius:10px;
    background:linear-gradient(90deg,#2563EB,#8B5CF6,#C084FC);
    box-shadow:0 0 10px rgba(139,92,246,0.55);
"></div>

<p style="
    color:#374151;
    font-size:15px;
    line-height:1.8;
">
Human language is naturally messy and inconsistent.
Before a machine learning model can represent text numerically,
the raw text must first be
<b style="color:#4F46E5;">cleaned and standardized</b>.
</p>

<div style="
    background:linear-gradient(120deg,#E8EEFF,#F3E8FF);
    border:1px solid #C4B5FD;
    border-radius:14px;
    padding:18px 20px;
    margin:18px 0;
    box-shadow:inset 0 1px 0 rgba(255,255,255,0.9);
">

<p style="
    margin:0 0 9px 0;
    color:#6D28D9;
    font-weight:700;
    font-size:14px;
">
Same meaning — different forms
</p>

<p style="
    margin:0;
    font-size:20px;
    font-weight:700;
    color:#1F2937;
    letter-spacing:0.5px;
">
"Good" &nbsp;&nbsp; → &nbsp;&nbsp; "good" &nbsp;&nbsp; → &nbsp;&nbsp; "good!"
</p>

</div>

<p style="
    color:#374151;
    font-size:15px;
    line-height:1.8;
    margin-bottom:0;
">
Without preprocessing, these variations may be treated differently.
The goal is therefore to
<b style="color:#7C3AED;">reduce unnecessary variation</b>
while preserving the information that actually matters for the task.
</p>

</div>

<div style="
    font-family:'Segoe UI', Arial, sans-serif;
    background:linear-gradient(135deg,#F7F9FF 0%,#EEF3FF 45%,#F8F3FF 100%);

STEP 01
</p>

<h2 style="
    margin:0;
    color:#312E81;

    font-weight:800;
">
2. Tokenization
</h2>

<div style="
    width:72px;
    height:4px;
    margin:14px 0 20px 0;
    border-radius:10px;
    background:linear-gradient(90deg,#2563EB,#8B5CF6,#C084FC);
    box-shadow:0 0 10px rgba(139,92,246,0.55);
"></div>

<p style="
    color:#374151;
    font-size:15px;
    line-height:1.8;
">
<b style="color:#4F46E5;">Tokenization</b> is the process of splitting
raw text into smaller units called
<b style="color:#7C3AED;">tokens</b>.
These tokens may be words, sub-words, or characters depending on the tokenizer.
</p>

<div style="
    background:linear-gradient(120deg,#E9EFFF,#F5EAFF);
    border:1px solid #C4B5FD;
    border-radius:14px;
    padding:18px 20px;
    margin-top:17px;
    box-shadow:inset 0 1px 0 rgba(255,255,255,0.9);
">

<p style="margin:0; color:#374151; font-size:15px;">
<b style="color:#4338CA;">Raw Text</b><br>
"The movie was great!"
</p>

<div style="
    text-align:center;
    color:#7C3AED;
    font-size:24px;
    margin:8px 0;
">
↓
</div>

<p style="margin:0; color:#374151; font-size:15px;">
<b style="color:#4338CA;">Tokens</b><br>
["The", "movie", "was", "great", "!"]
</p>

</div>

<p style="
    color:#4B5563;
    font-size:14px;
    line-height:1.7;
    margin:17px 0 0 0;
">
Modern transformer models often use
<b style="color:#7C3AED;">sub-word tokenization</b>,
which allows unfamiliar words to be divided into smaller known pieces.
</p>

</div>

<div style="
    font-family:'Segoe UI', Arial, sans-serif;
    background:linear-gradient(135deg,#F8FAFF,#F1F5FF,#FAF5FF);

STEP 02

<h2 style="

    color:#312E81;

">
3. Cleaning & Normalization
</h2>

<div style="
    width:72px;
    height:4px;
    margin:14px 0 20px 0;
    border-radius:10px;
    background:linear-gradient(90deg,#2563EB,#8B5CF6,#C084FC);
    box-shadow:0 0 10px rgba(139,92,246,0.55);
"></div>

<p style="color:#374151; font-size:15px; line-height:1.8;">
Raw text often contains inconsistent capitalization, punctuation,
and other elements that unnecessarily increase the vocabulary.
<b style="color:#4F46E5;">Cleaning and normalization</b>
reduce these inconsistencies before numerical representation.
</p>

<div style="
    background:linear-gradient(120deg,#E9EFFF,#F5EAFF);
    border:1px solid #C4B5FD;
    border-radius:14px;
    padding:18px 20px;
    margin-top:18px;
">

<table style="width:100%; font-size:15px; color:#374151;">
<tr>
<td><b style="color:#4338CA;">Lowercasing</b></td>
<td>Good → good</td>
</tr>

<tr>
<td style="padding-top:12px;"><b style="color:#7C3AED;">Punctuation Removal</b></td>
<td style="padding-top:12px;">great! → great</td>
</tr>
</table>

</div>

<p style="color:#4B5563; font-size:14px; line-height:1.7; margin-bottom:0;">
Cleaning should remove unnecessary variation without destroying
information that may be important for the prediction task.
</p>

</div>

<div style="
    font-family:'Segoe UI', Arial, sans-serif;
    background:linear-gradient(135deg,#F8FAFF,#F1F5FF,#FAF5FF);

STEP 03

<h2 style="
    margin:0;
    color:#312E81;
    font-size:25px;
    font-weight:800;
">
4. Stop-word Removal
</h2>

<div style="
    width:90px;
    height:5px;
    margin:14px 0 24px 0;
    border-radius:10px;
    background:linear-gradient(90deg,#2563EB,#8B5CF6,#C084FC);
    box-shadow:0 0 12px rgba(139,92,246,0.6);
"></div>

<p style="
    color:#475569;
    font-size:15px;
    line-height:1.9;
">
<b style="color:#4F46E5;">Stop words</b> are very common words such as
"the", "is", and "a". In some NLP tasks, these words add little useful
information, so removing them can reduce noise and simplify the text.
</p>

<div style="
    background:#F8FAFC;
    border:1px solid #E2E8F0;
    border-radius:14px;
    padding:20px 22px;
    margin-top:20px;
">

<h3 style="
    color:#334155;
    font-size:21px;
    font-weight:500;
    margin:0 0 16px 0;
">
What Can Go Wrong?
</h3>

<p style="
    color:#475569;
    font-size:15px;
    line-height:2;
    margin:0;
">
<b style="color:#7C3AED;">Original:</b>
&nbsp; "The movie is not good."<br>

<b style="color:#DB2777;">Wrong removal:</b>
&nbsp; "movie good"<br>

<b style="color:#059669;">Safer result:</b>
&nbsp; "movie not good"
</p>

</div>

<div style="
    background:#FFF1F2;
    border-left:6px solid #EF4444;
    border-radius:12px;
    padding:17px 20px;
    margin-top:18px;
">

<p style="
    color:#991B1B;
    font-size:15px;
    line-height:1.8;
    margin:0;
">
<b>Key idea:</b>
Stop-word removal is task-dependent. Negation words such as
<b>not</b>, <b>no</b>, and <b>nor</b> may completely change the meaning,
so they should not be removed blindly.
</p>

</div>

</div>

<div style="
    font-family:'Segoe UI', Arial, sans-serif;

<h2 style="
    margin:0;
    color:#312E81;
    font-size:25px;
    font-weight:800;
">
5. Lemmatization
</h2>

<div style="
    width:90px;
    height:5px;
    margin:14px 0 24px 0;
    border-radius:10px;
    background:linear-gradient(90deg,#2563EB,#8B5CF6,#C084FC);
    box-shadow:0 0 12px rgba(139,92,246,0.6);
"></div>

<p style="
    color:#475569;
    font-size:15px;
    line-height:1.9;
">
After cleaning the text, different grammatical forms of the same word may
still remain. <b style="color:#4F46E5;">Lemmatization</b> reduces these
forms to their meaningful dictionary base form, called a
<b style="color:#7C3AED;">lemma</b>.
</p>

<div style="
    background:#F8FAFC;
    border:1px solid #E2E8F0;
    border-radius:14px;
    padding:20px 22px;
    margin-top:20px;
">

<h3 style="
    color:#334155;
    font-size:21px;
    font-weight:500;
    margin:0 0 16px 0;
">
Examples
</h3>

<p style="
    color:#475569;
    font-size:16px;
    line-height:2.1;
    margin:0;
">

<b style="color:#7C3AED;">movies</b>
&nbsp; → &nbsp;
<b style="color:#059669;">movie</b><br>

<b style="color:#DB2777;">actors</b>
&nbsp; → &nbsp;
<b style="color:#059669;">actor</b><br>

<b style="color:#2563EB;">running</b>
&nbsp; → &nbsp;
<b style="color:#059669;">run</b>

</p>

</div>

<div style="
    background:#FFF7ED;
    border-left:6px solid #F59E0B;
    border-radius:12px;
    padding:17px 20px;
    margin-top:18px;
">

<p style="
    color:#92400E;
    font-size:15px;
    line-height:1.8;
    margin:0;
">
<b>Lemmatization vs. Stemming:</b>
Stemming may simply cut word endings, while lemmatization aims to return
a valid dictionary form. This usually makes the resulting text easier
to interpret.
</p>

</div>

<h3 style="color:#BE185D;font-size:17px;font-weight:650;margin:0 0 5px 0;">3.1 Cleaning & Normalization</h3>
<p style="color:#475569;font-size:14px;line-height:1.7;margin:0;">Decode HTML, remove HTML tags, normalize case, remove numbers and unnecessary punctuation, and collapse repeated whitespace.</p>
</div>

```python
import re
import html

cleaned_review = html.unescape(sample_review)
cleaned_review = re.sub(r"<[^>]+>", " ", cleaned_review)
cleaned_review = cleaned_review.lower()
cleaned_review = re.sub(r"\d+", " ", cleaned_review)
cleaned_review = re.sub(r"[^\w\s']", " ", cleaned_review)
cleaned_review = re.sub(r"\s+", " ", cleaned_review).strip()

print("RAW REVIEW:")
print(sample_review[:500])

print("\nCLEANED REVIEW:")
print(cleaned_review[:500])
```

<h3 style="color:#BE185D;font-size:17px;font-weight:650;margin:0 0 5px 0;">3.2 Stop-word Removal</h3>
<p style="color:#475569;font-size:14px;line-height:1.7;margin:0;">Remove common low-information words while deliberately preserving <b>not</b>, <b>no</b>, and <b>nor</b> because they can change sentiment.</p>
</div>

```python
from nltk.corpus import stopwords

nltk.download("stopwords")

stop_words = set(stopwords.words("english"))
negation_words = {"not", "no", "nor"}
safe_stop_words = stop_words - negation_words

tokens_without_stopwords = [
    token for token in cleaned_tokens
    if token not in safe_stop_words
]

print("FIRST 80 TOKENS BEFORE STOP-WORD REMOVAL:")
print(cleaned_tokens[:80])

print("\nFIRST 80 TOKENS AFTER STOP-WORD REMOVAL:")
print(tokens_without_stopwords[:80])

print("\nTOKENS BEFORE STOP-WORD REMOVAL:", len(cleaned_tokens))
print("TOKENS AFTER STOP-WORD REMOVAL:", len(tokens_without_stopwords))

print("\nNEGATION CHECK:")
for word in negation_words:
    if word in cleaned_tokens:
        print(f"'{word}' preserved:", word in tokens_without_stopwords)
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from nltk import pos_tag

nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("averaged_perceptron_tagger_eng")

lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    if tag.startswith("V"):
        return wordnet.VERB
    if tag.startswith("N"):
        return wordnet.NOUN
    if tag.startswith("R"):
        return wordnet.ADV
    return wordnet.NOUN

pos_tags = pos_tag(tokens_without_stopwords)

lemmatized_tokens = [
    lemmatizer.lemmatize(word, get_wordnet_pos(tag))
    for word, tag in pos_tags
]

print("FIRST 80 TOKENS BEFORE LEMMATIZATION:")
print(tokens_without_stopwords[:80])

print("\nFIRST 80 TOKENS AFTER LEMMATIZATION:")
print(lemmatized_tokens[:80])
```

<h3 style="color:#047857;font-size:17px;font-weight:650;margin:0 0 5px 0;">5.1 Build the Reusable Function</h3>
<p style="color:#475569;font-size:14px;line-height:1.7;margin:0;">Encapsulate cleaning, tokenization, safe stop-word removal, POS-aware lemmatization, and contraction refinement in one function.</p>
</div>

```python
def preprocess_review(text):
    text = html.unescape(text)
    text = re.sub(r"<[^>]+>", " ", text)
    text = text.lower()
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"[^\w\s']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    tokens = word_tokenize(text)

    safe_stop_words = stop_words - {"not", "no", "nor", "n't"}
    tokens = [token for token in tokens if token not in safe_stop_words]

    tagged_tokens = pos_tag(tokens)

    tokens = [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tagged_tokens
    ]

    refined_tokens = []
    for token in tokens:
        if token == "n't":
            refined_tokens.append("not")
        elif token.startswith("'"):
            continue
        else:
            refined_tokens.append(token)

    return " ".join(refined_tokens)
```
