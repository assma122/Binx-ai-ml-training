<div style="font-size:22px;font-weight:800;color:#1E293B;margin-bottom:8px;">
What Does <span style="color:#7C3AED;">Model Integration</span> Mean?
</div>

<div style="font-size:13px;line-height:1.75;color:#475569;">
During experimentation, the machine learning workflow is often divided into separate steps:
<span style="color:#2563EB;font-weight:700;">text cleaning</span>,
<span style="color:#7C3AED;font-weight:700;">feature extraction</span>,
<span style="color:#DB2777;font-weight:700;">model prediction</span>,
and
<span style="color:#059669;font-weight:700;">evaluation</span>.
<br><br>

<span style="color:#7C3AED;font-weight:700;">Model integration</span>
connects these separate pieces into <span style="color:#0F766E;font-weight:700;">one clear and consistent flow</span>,
so that a raw movie review can enter the system and a final sentiment prediction comes out.

</div>

<b style="color:#B45309;">💡 Why do we need integration?</b>

<div style="color:#475569;font-size:13px;line-height:1.75;margin-top:6px;">
Instead of manually running several notebook cells every time we want a prediction,
<span style="color:#7C3AED;font-weight:700;">integration</span>
allows us to build
<span style="color:#059669;font-weight:700;">one organized prediction process</span>.
<br><br>

Later, one <code style="color:#BE185D;background:#FDF2F8;padding:2px 6px;border-radius:6px;">predict()</code>
function will receive a <span style="color:#2563EB;font-weight:700;">raw IMDb review</span>
and automatically apply all required steps before returning the <span style="color:#059669;font-weight:700;">final sentiment</span>.

</div>

<div style="font-size:13px;line-height:1.75;color:#475569;margin-top:7px;">
<span style="color:#7C3AED;font-weight:700;">Integration</span>
does <b>not</b> mean creating a new model.
<br>
It means connecting the
<span style="color:#2563EB;font-weight:700;">preprocessing</span>
and the
<span style="color:#DB2777;font-weight:700;">already-trained model</span>

<b style="color:#1E293B;">Input:</b> <span style="color:#2563EB;font-weight:700;">Raw movie review</span>
  →   <b style="color:#1E293B;">Output:</b> <span style="color:#059669;font-weight:700;">Positive or Negative sentiment</span>

<div style="
    font-size:13px;

">

<div style="

<span style="color:#059669;font-weight:700;">identical</span>
to the preprocessing used during training.
</div>

<div style="
    font-size:13px;
    line-height:1.55;
    color:#475569;

How Can Skew Happen?
</div>

<div style="
    font-size:13px;
    line-height:1.55;
    color:#475569;
">
In our IMDb model, skew can happen if new reviews are cleaned differently from the reviews used during training.
</div>

<div style="
    font-size:13px;
    line-height:1.55;
    color:#475569;
    margin-top:7px;
">
It can also happen if we create a
<span style="color:#DB2777;font-weight:700;">new TF-IDF vectorizer</span>
for incoming reviews instead of reusing the vectorizer already fitted on the training data.
</div>

<div style="
    font-size:13px;
    line-height:1.55;
    color:#475569;
    margin-top:7px;
">
In both cases, the model receives a different representation from the one it learned from, which may lead to
<span style="color:#BE123C;font-weight:800;">incorrect predictions</span>.
</div>

```python
import re
import html
import nltk

from nltk.corpus import stopwords, wordnet
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

# Required NLTK resources
resources = [
    "punkt",
    "punkt_tab",
    "stopwords",
    "wordnet",
    "omw-1.4",
    "averaged_perceptron_tagger_eng"
]

for resource in resources:
    nltk.download(resource, quiet=True)

# Preserve sentiment-critical negation
negation_words = {"not", "no", "nor"}

stop_words = set(stopwords.words("english")) - negation_words
lemmatizer = WordNetLemmatizer()


def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("N"):
        return wordnet.NOUN
    elif tag.startswith("R"):
        return wordnet.ADV
    else:
        return wordnet.NOUN


def preprocess_review(text):
    # Decode HTML entities and remove HTML tags
    text = html.unescape(str(text))
    text = re.sub(r"<[^>]+>", " ", text)

    # Preserve contractions such as "isn't" -> "is not"
    text = re.sub(r"n['’]t\b", " not", text)

    # Lowercase
    text = text.lower()

    # Remove numbers and punctuation
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"[^\w\s']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenize
    tokens = word_tokenize(text)

    # Remove stop words while preserving negation
    tokens = [
        token for token in tokens
        if token not in stop_words and len(token) > 1
    ]

    # POS-aware lemmatization
    tagged_tokens = pos_tag(tokens)

    lemmas = [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tagged_tokens
    ]

    # Remove remaining apostrophe fragments
    lemmas = [
        word for word in lemmas
        if not word.startswith("'")
    ]

    return " ".join(lemmas)


print("Preprocessing pipeline ready.")
```

```python
import pandas as pd
from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Load the same IMDb dataset
data_path = Path("../Day1/IMDB Dataset.csv")

df = pd.read_csv(data_path)

# Same duplicate-removal decision used earlier
df = df.drop_duplicates().reset_index(drop=True)

# Apply the same preprocessing
df["processed_review"] = df["review"].apply(preprocess_review)

# Convert labels to numeric values
df["label"] = df["sentiment"].map({
    "negative": 0,
    "positive": 1
})


# Split the data while keeping the original reviews
train_df, test_df = train_test_split(
    df[["review", "processed_review", "label"]],
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)

X_train = train_df["processed_review"]
X_test = test_df["processed_review"]

y_train = train_df["label"]
y_test = test_df["label"]
```
