```python
from pathlib import Path
import urllib.request
import tarfile

dataset_url = "https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz"

data_dir = Path("flower_dataset")
archive_path = Path("flower_photos.tgz")

if not data_dir.exists():
    print("Downloading Flowers dataset...")

    urllib.request.urlretrieve(dataset_url, archive_path)

    with tarfile.open(archive_path, "r:gz") as tar:
        tar.extractall(".")

    Path("flower_photos").rename(data_dir)
    archive_path.unlink()

    print("Dataset downloaded successfully.")
else:
    print("Dataset already exists.")

class_names = sorted([
    folder.name
    for folder in data_dir.iterdir()
    if folder.is_dir()
])

print("Dataset path:", data_dir)
print("Classes:", class_names)
print("Number of classes:", len(class_names))
```

```python
import pandas as pd

image_extensions = {".jpg", ".jpeg", ".png"}

class_counts = {}

for class_name in class_names:
    class_dir = data_dir / class_name

    image_files = [
        path for path in class_dir.iterdir()
        if path.suffix.lower() in image_extensions
    ]

    class_counts[class_name] = len(image_files)

inventory_df = pd.DataFrame({
    "class": class_counts.keys(),
    "image_count": class_counts.values()
})

total_images = inventory_df["image_count"].sum()

print("Total images:", total_images)
inventory_df
```

```python
import cv2
import matplotlib.pyplot as plt
import random

random.seed(42)

sample_images = []

for class_name in class_names:
    class_dir = data_dir / class_name

    image_files = sorted([
        path for path in class_dir.iterdir()
        if path.suffix.lower() in image_extensions
    ])

    image_path = random.choice(image_files)

    image_bgr = cv2.imread(str(image_path))
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

    sample_images.append(
        (class_name, image_path, image_bgr, image_rgb)
    )

fig, axes = plt.subplots(1, 5, figsize=(16, 4))

for ax, (class_name, image_path, image_bgr, image_rgb) in zip(axes, sample_images):
    height, width = image_bgr.shape[:2]

    ax.imshow(image_rgb)
    ax.set_title(
        f"{class_name.title()}\n{width} × {height}",
        fontsize=10,
        fontweight="bold"
    )
    ax.axis("off")

plt.suptitle(
    "Raw Flowers Dataset Preview",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
image_sizes = []

for class_name in class_names:
    class_dir = data_dir / class_name

    for image_path in class_dir.iterdir():
        if image_path.suffix.lower() in image_extensions:
            image = cv2.imread(str(image_path))

            if image is not None:
                height, width = image.shape[:2]

                image_sizes.append({
                    "class": class_name,
                    "width": width,
                    "height": height
                })

sizes_df = pd.DataFrame(image_sizes)

plt.figure(figsize=(9, 6))

for class_name in class_names:
    class_data = sizes_df[sizes_df["class"] == class_name]

    plt.scatter(
        class_data["width"],
        class_data["height"],
        alpha=0.45,
        s=18,
        label=class_name
    )

plt.axvline(224, linestyle="--", linewidth=1.2)
plt.axhline(224, linestyle="--", linewidth=1.2)

plt.xlabel("Image Width")
plt.ylabel("Image Height")
plt.title(
    "Image Dimension Distribution Before Resizing",
    fontsize=14,
    fontweight="bold"
)

plt.legend(title="Class")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
def preprocess_image(image_path, target_size=(224, 224)):
    image_bgr = cv2.imread(str(image_path))

    if image_bgr is None:
        raise ValueError(f"Could not read image: {image_path}")

    resized_bgr = cv2.resize(image_bgr, target_size)
    image_rgb = cv2.cvtColor(resized_bgr, cv2.COLOR_BGR2RGB)

    normalized_image = image_rgb.astype(np.float32) / 255.0

    return normalized_image
```

```python
augmentation_input = np.expand_dims(
    correct_rgb.astype(np.float32),
    axis=0
)

augmentation_flow = datagen.flow(
    augmentation_input,
    batch_size=1,
    shuffle=False
)

fig, axes = plt.subplots(2, 3, figsize=(12, 8))

for i, ax in enumerate(axes.flat):
    augmented = next(augmentation_flow)[0]

    augmented = np.clip(
        augmented,
        0,
        255
    ).astype(np.uint8)

    ax.imshow(augmented)
    ax.set_title(
        f"Augmented View {i + 1}",
        fontsize=10,
        fontweight="bold"
    )
    ax.axis("off")

plt.suptitle(
    "One Image, Multiple Training Variations",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

<div style="font-size:20px;font-weight:800;color:#1E293B;margin-bottom:5px;">Verify Both Model Paths</div>

<div style="font-size:12.5px;line-height:1.6;color:#475569;">Create one reusable function for a custom model and another for MobileNetV2, then verify that both preserve the same geometry while producing the numerical range expected by each model.</div>

<div style="margin-top:10px;padding:10px 12px;border-radius:10px;background:#FFFFFF;border:1px solid #DDD6FE;font-size:11.5px;color:#64748B;">
<b style="color:#6D28D9;">Validation target:</b> same 224 × 224 × 3 shape · different preprocessing contract.

```python
def prepare_for_custom_model(image_path, target_size=(224, 224)):
    image_bgr = cv2.imread(str(image_path))

    if image_bgr is None:
        raise ValueError(f"Could not read image: {image_path}")

    image_bgr = cv2.resize(image_bgr, target_size)
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

    return image_rgb.astype(np.float32) / 255.0


def prepare_for_mobilenet_v2(image_path, target_size=(224, 224)):
    image_bgr = cv2.imread(str(image_path))

    if image_bgr is None:
        raise ValueError(f"Could not read image: {image_path}")

    image_bgr = cv2.resize(image_bgr, target_size)
    image_rgb = cv2.cvtColor(
        image_bgr,
        cv2.COLOR_BGR2RGB
    ).astype(np.float32)

    return preprocess_input(image_rgb)
```

```python
custom_input = prepare_for_custom_model(target_path)
mobilenet_ready = prepare_for_mobilenet_v2(target_path)

final_validation = pd.DataFrame({
    "Pipeline": [
        "Custom Model",
        "MobileNetV2"
    ],
    "Shape": [
        str(custom_input.shape),
        str(mobilenet_ready.shape)
    ],
    "Data Type": [
        str(custom_input.dtype),
        str(mobilenet_ready.dtype)
    ],
    "Minimum": [
        round(float(custom_input.min()), 3),
        round(float(mobilenet_ready.min()), 3)
    ],
    "Maximum": [
        round(float(custom_input.max()), 3),
        round(float(mobilenet_ready.max()), 3)
    ]
})

final_validation
```
