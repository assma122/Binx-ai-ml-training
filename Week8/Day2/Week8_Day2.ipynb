Bag-of-Words and TF-IDF
</div>

<div style="
font-size:15px;
line-height:1.7;
color:#475569;
margin-bottom:15px;
">
Bag-of-Words represents a document by counting which words appear and how often they occur, while ignoring word order.
TF-IDF improves this idea by giving each word a weight based on how important it is in one document compared with the full collection of documents.
</div>

<div style="
padding:15px 17px;
background:#FFFFFF;
border-radius:12px;
border:1px solid #FDE7C2;
margin-bottom:12px;
">

<div style="
font-size:14px;
font-weight:700;
color:#B45309;
margin-bottom:8px;
">
TF — Term Frequency
</div>

<div style="
font-size:14px;
line-height:1.6;
color:#475569;
">
Measures how often a word appears inside a document. A word that appears repeatedly in one review receives a stronger local signal.
</div>

</div>

<div style="
padding:15px 17px;
background:#FFFFFF;
border-radius:12px;
border:1px solid #FDE7C2;
margin-bottom:12px;
">

<div style="
font-size:14px;
font-weight:700;
color:#B45309;
margin-bottom:8px;
">
IDF — Inverse Document Frequency
</div>

<div style="
font-size:14px;
line-height:1.6;
color:#475569;
">
Reduces the importance of words that appear in many documents and increases the importance of words that are more distinctive.
</div>

</div>

<div style="
padding:15px 17px;
background:#FFF9ED;

">

<div style="

">
Example
</div>

<div style="
font-size:14px;
line-height:1.7;
color:#475569;
">
A common word such as <b>movie</b> may appear in thousands of IMDb reviews, so its TF-IDF weight can be relatively low.
A more distinctive word such as <b>masterpiece</b> may appear in fewer reviews, so it can receive a higher weight and help distinguish sentiment.
</div>

</div>

<div style="
margin-top:15px;
font-size:13px;
line-height:1.6;
color:#64748B;
">
TF-IDF is fast, interpretable, and often a very strong baseline for text classification. Its main limitation is that it treats words independently and does not understand their semantic meaning.
</div>

</div>

</div>

<div style="
font-size:15px;
line-height:1.7;
color:#475569;
margin-bottom:15px;
">
Word embeddings represent each word as a dense numerical vector. Instead of describing only how often a word appears, the vector is positioned so that words with similar meanings are located close to each other in vector space.
</div>

<div style="

</div>

<div style="

">
Words such as <b>good</b>, <b>great</b>, and <b>excellent</b> should have vectors that are closer to each other than to unrelated words such as <b>camera</b> or <b>river</b>.

</div>

<div style="

<b>Word2Vec</b> and <b>GloVe</b> are classic embedding methods. They learn word representations from patterns of co-occurrence in large text collections.

</div>

<div style="

The vector itself may contain hundreds of numeric dimensions, but the important idea is the <b>relative position</b>: words used in similar contexts tend to become neighbors in the embedding space.

Unlike TF-IDF, embeddings can capture semantic relationships between words. However, classic embeddings still assign one fixed vector to each word, regardless of the sentence in which it appears.

<div style="

```python
import re
import html
import nltk

from nltk.corpus import stopwords, wordnet
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

# Download required NLTK resources
resources = [
    "punkt",
    "punkt_tab",
    "stopwords",
    "wordnet",
    "omw-1.4",
    "averaged_perceptron_tagger_eng"
]

for resource in resources:
    nltk.download(resource, quiet=True)

# Preserve sentiment-critical negation
negation_words = {"not", "no", "nor"}

stop_words = set(stopwords.words("english"))
stop_words = stop_words - negation_words

lemmatizer = WordNetLemmatizer()


def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("N"):
        return wordnet.NOUN
    elif tag.startswith("R"):
        return wordnet.ADV
    else:
        return wordnet.NOUN


def preprocess_review(text):
    # Decode HTML entities and remove HTML tags
    text = html.unescape(str(text))
    text = re.sub(r"<[^>]+>", " ", text)

    # Normalize contractions before removing punctuation
    text = re.sub(r"n['’]t\b", " not", text)

    # Lowercase
    text = text.lower()

    # Remove numbers and punctuation
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"[^\w\s']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenize
    tokens = word_tokenize(text)

    # Remove stop words while preserving negation
    tokens = [
        token for token in tokens
        if token not in stop_words and len(token) > 1
    ]

    # POS-aware lemmatization
    tagged_tokens = pos_tag(tokens)

    lemmas = [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tagged_tokens
    ]

    # Remove remaining apostrophe fragments
    lemmas = [
        word for word in lemmas
        if not word.startswith("'")
    ]

    return " ".join(lemmas)


print("Preprocessing pipeline ready.")
```

These relationships reflect similarity in language usage and context rather than exact synonymy.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

semantic_words = [
    "movie", "film", "cinema",
    "actor", "actress", "director",
    "good", "great", "excellent",
    "bad", "worse", "terrible",
    "love", "loved", "hate"
]

word_vectors = np.array([
    embedding_model[word]
    for word in semantic_words
])

pca = PCA(n_components=2)

vectors_2d = pca.fit_transform(word_vectors)

plt.figure(figsize=(10, 7))

plt.scatter(
    vectors_2d[:, 0],
    vectors_2d[:, 1],
    s=70
)

for i, word in enumerate(semantic_words):
    plt.annotate(
        word,
        (vectors_2d[i, 0], vectors_2d[i, 1]),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=10
    )

plt.title("Semantic Geometry of GloVe Word Embeddings")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
