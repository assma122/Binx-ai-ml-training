## import numpy 

```python
import numpy as np 
```

## Vectors 
A vector is a one-dimensional array ,it can represent one student with different feature 

```python
student = np.array([20,6,90]) ## age = 20 , study hours = 6 , exam score =90
print(student)
print("shape:",student.shape)
```

## Matrices 
A matrix is a group of vectors 
each row represents one student , and each column represents one feature 

```python
student = np.array([
    [20,6,90],
                   [22,4,85],
                   [21,8,95]])
print(student)
print("shape",student.shape)
```

the matrix contains 3 students 
columns : age,study hours , exam score .
The shape (3,3)

## Dot product 
The dot product (or scalar product) of two vectors is a way of multiplying vectors that results in a scalar (number). This operation, often symbolized by a centered dot, is dependent on the length of both vectors and the angle between them

```python
a = np.array([4,6,9])
b = np.array([2,3,4])
result = np.dot(a,b)
print(result)
```

## Matrix Multiplication 
 multiplication can be used to calculate results for more than one student at the same time

## RULE 
(M X N) X (N X P)

```python
Student =  np.array([[4,80],[7,60],[8,95]])
w = np.array([2,1.5])
r = Student@w
print(r)
```
