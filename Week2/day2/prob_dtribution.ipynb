# Day 2 - Probability and Distributions

Probability is used to measure how likely an event is to happen.

Its value is between 0 and 1, where 0 means the event is impossible and 1 means the event is certain.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Probability Basics

Probability can be calculated by dividing the number of favorable outcomes by the total number of possible outcomes.

For example, the probability of getting heads when flipping a fair coin is 1 out of 2.

```python
heads = 1
total = 2

probability = heads / total

print("probability of head:", probability)
```

The probability of getting heads is 0.5, which means there is a 50% chance.

## Complement Rule

The complement rule gives the probability that an event does not happen.

It is calculated by subtracting the probability of the event from 1.

```python
pr_heads = 0.5

pr_not_heads = 1 - pr_heads

print("probability of not getting heads:", pr_not_heads)
```

The probability of not getting heads is 0.5 because the only other result is tails.

## Addition Rule

The addition rule is used to find the probability of event A or event B
is the one for inclusive OR: 

P(A∪B)=P(A)+P(B)−P(A∩B).

```python
pa = 0.3
pb = 0.4
p_a_and_b = 0.1

p_a_or_b = pa + pb - p_a_and_b

print("Probability of A or B:", p_a_or_b)
```

## Multiplication Rule

The multiplication rule is used to find the probability of two independent events happening together.

```python
p_first= 0.5
p_second= 0.5

p_both_heads = p_first * p_second

print("Probability of getting two heads:", p_both_heads)
```

The probability of getting heads two times is 0.25 because the two coin flips are independent.

## Conditional Probability

Conditional probability is the probability of event A happening when event B has already happened.

```python
students_with_course = 40
students_with_course_and_passed = 30

p = students_with_course_and_passed / students_with_course
print("Conditional probability:", p)
```

The probability is 0.75, which means that 75% of the students who completed the course passed.

### Bayes' Theorem 

I used Bayes' theorem to find the chance that a student studies every day after knowing that the student passed.

```python
study = 0.30
pass_if_study = 0.70
passs = 0.50
result = (pass_if_study *study)/passs
print(result )
```

which means 42% of the students who passed may study every day 

## Common Distributions

A probability distribution shows how values are spread.

The main distributions are normal, binomial, and uniform.

### Normal Distribution

The normal distribution has a bell shape, and most values are close to the mean.

```python
normal = np.random.normal(40, 10, 100)

print("Mean:", np.mean(normal))
print("Std:", np.std(normal))
```

The values are close to a mean of 50 and a standard deviation of 10.

### Binomial Distribution

The binomial distribution counts the number of successes in repeated trials.

```python
success = np.random.binomial(20,0,7)
print(success)
```

This code gives the number of successes in 20 trials.

### Uniform Distribution

In a uniform distribution, all values in the range have the same chance.

```python
v = np.random.uniform(15,40,10)
print(v) # the code generated 10 rand values btw 15 and 40
```

### Conditional Probability Simulation

I used a simple example to compare the manual probability with the simulation result.

```python
np.random.seed(1)
student = 1000
joinedCourse = np.random.rand(student)<0.4
passed = joinedCourse&(np.random.rand(student)<0.75)
joinedCount = np.sum(joinedCourse)
pass_c = np.sum(passed)
r = pass_c/joinedCount
print(r)
```

The simulation result is close to 0.75, so it agrees with the manual probability.
