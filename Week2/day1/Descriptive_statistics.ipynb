# Day 1 - Descriptive Statistics

Descriptive statistics are used to summarize and understand the main features of a dataset.

They help describe the center of the data using the mean, median, and mode, and describe the spread using the range, variance, standard deviation, and IQR.

## Measures of Central Tendency

Measures of central tendency describe the center of the data.

The main measures are mean, median, and mode.

### Mean

The mean is the average of all values.

It is useful when the data does not have extreme outliers.

```python
import numpy as np
import pandas as pd
```

### Median 
the median is the middle value after sorting the data 


### Mode 
the mode is the value that appears most often in the data 

```python
data = np.array([15, 8, 15, 7, 92])
mean_value = np.mean(data)
median_value = np.median(data)
mode_value = pd.Series(data).mode()[0]

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

the mean is affected by the outlier while the median close to the normal val 

## Measures of Spread

Measures of spread describe how similar or varied the set of observed values are for a particular variable (data item). Measures of spread include the range, quartiles and the interquartile range, variance and standard deviation.

### Range

The range is the difference between the maximum and minimum values.

```python
R = max(data)-min(data)
print(R)
```

### Variance
show how far the values are spread from the mean 

```python
v = np.var(data)
print(v)
```

### Standard Deviation
 shows the spread of the value around the mean

```python
standard = np.std(data)
print(standard)
```

### Quartiles and IQR

- Q1 and Q3 divide the data into parts
- The IQR shows the spread of the middle 50% of the data

```python
q1 = np.percentile(data,26)
q3 =np.percentile(data,75)
print(q1)
print(q3)
```

```python
iqr = q3 - q1 
print(iqr)
```

### Observation

* The value 92 makes the range, variance, and standard deviation higher
 * The IQR is less affected because it focuses on the middle values.
