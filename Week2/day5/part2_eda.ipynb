# Day 5 - EDA Part 2: Correlation & Data Storytelling

Correlation analysis is used to measure the relationship between numeric variables.

It helps us understand how variables are related and identify patterns in the dataset.

## Step 1: Bivariate Analysis

Bivariate analysis is used to study the relationship between two variables.

In this step, I created scatter plots and grouped box plots to explore these relationships.

## Scatter Plot

Scatter plots show the relationship between two numeric variables.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("insurance.csv")

df.head()
```

```python
sns.scatterplot(data=df,x="age",y="charges")
plt.title("Age Vs Charges")
plt.show()

sns.scatterplot(data=df,x="bmi",y="charges")
plt.title("Bmi Vs Charges")
```

### Observation

- The scatter plot between age and charges shows a positive relationship. As age increases, insurance charges generally increase.
- The scatter plot between BMI and charges shows a weak positive relationship. Higher BMI values may be associated with higher charges, but the relationship is not very clear.
- Some points have very high charges compared to the rest of the data.

## Grouped Box Plot

Grouped box plots compare a numeric variable across different categories.

```python
sns.boxplot(data=df,x="smoker",y="charges")
plt.title("Charges Vs Smoker")
plt.show()

sns.boxplot(data=df,x="sex",y="charges")
plt.title("Charges Vs Sex")
plt.show()


sns.boxplot(data=df,x="region",y="charges")
plt.title("Charges Vs region")
plt.show()
```

### Observation

- Smokers have much higher insurance charges than non-smokers.
- The charges for males and females are similar, with no major difference.
- The charges are distributed similarly across the four regions, with only small differences.

## Correlation Matrix

In this step, I calculated the correlation between the numeric variables.

```python
corr = df.select_dtypes(include ="number").corr()
corr
```

## Correlation Heatmap

The heatmap helps visualize the strength of the relationships between numeric variables.

```python
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("corelation heatmap")
plt.show()
```

### Observation

- Age has the strongest positive correlation with charges, with a value of 0.30.
- BMI has a weak positive correlation with charges, with a value of 0.20.
- The number of children has a very weak positive correlation with charges, with a value of 0.068.
- The relationships between the other numeric variables are also weak.
- There are no negative correlations between the numeric variables.

## Pairplot

The pairplot shows the relationships between all numeric variables in one figure.

```python
sns.pairplot(df.select_dtypes(include="number"))

plt.show()
```

### Observation

- Age shows a positive relationship with charges, as charges generally increase with age.
- BMI has a weak positive relationship with charges, although the data points are widely scattered.
- Children has no clear relationship with charges.
- Most numeric variables have weak relationships with each other.
- The distribution of charges is right-skewed, with most values concentrated at lower charges and a few very high values.

## Data Storytelling

### Key Findings

- Age has the strongest positive relationship with charges, with a correlation of 0.30.
- BMI has a weak positive relationship with charges, with a correlation of 0.20.
- The number of children has a very weak relationship with charges, with a correlation of 0.068.
- Most numeric variables have weak correlations.
- Insurance charges are right-skewed, with some customers having much higher charges than others.
- Age and BMI may be useful features for predicting insurance charges.
