## Introduction

Supervised learning is a type of machine learning where the model learns from labeled data. Each training example includes input features and the correct target value, allowing the model to learn patterns and make predictions for new, unseen data.

## What is Supervised Learning?

Supervised learning is a type of machine learning where the model learns from labeled data. Each training example contains input features and the correct target value. The model learns the relationship between the features and the target so it can make predictions for new, unseen data.

### Observation

Supervised learning uses labeled data to learn the relationship between the input features and the target. The trained model can then predict the target for new data.

## Regression vs. Classification

Supervised learning problems are divided into two main types: regression and classification. The difference depends on the type of target value that the model predicts.

### Observation

- Regression predicts continuous numeric values.
- Classification predicts categories or classes.
- The type of the target variable determines whether the problem is regression or classification.

## The Scikit-learn API

Scikit-learn uses the same workflow for almost all machine learning models.

```python
model = Model()
```

## Step 2 : Perform an 80/20 train/test split with a fixed random_state.

## Train/Test Split

The dataset is divided into training data and testing data.

The training set is used to teach the model.

The test set is kept separate and is used later to evaluate the model on unseen data.

````python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
````

the dataset is divided into training sets , 
80% of the data is used for training and 20% is kept for testing 
the test set remains unseen during training so the model can be evaluated fairly on new data

## Step 4 : Why Must the Model Not See the Test Set During Training?



The model must not see the test set during training because the test data is used to evaluate its performance on unseen examples.

If the model trains on the test set, it may memorize the answers and produce misleading results.

Keeping the test set separate gives a more honest evaluation of how the model may perform on new data.
