## Introduction

Logistic Regression is a supervised machine learning algorithm used for classification problems. It predicts the probability that a sample belongs to a specific class and then classifies it into one of the possible categories.

## Objective

The goal of this notebook is to train a Logistic Regression classifier, generate predictions, evaluate its performance using classification metrics, and interpret the results.

## What is Logistic Regression?

Logistic Regression is a supervised learning algorithm used for classification tasks.

Unlike Linear Regression, which predicts continuous numerical values, Logistic Regression predicts class labels such as 0 or 1.

The model calculates a weighted sum of the input features and then applies the sigmoid function to convert the result into a probability between 0 and 1.

## Dataset

I used the Pima Indians Diabetes Dataset.

The target variable is **Outcome**.

- Outcome = 0 → No Diabetes
- Outcome = 1 → Diabetes

## Hands - on lab : Building a Classifier 



##  Select Features and Target

The target variable is **Outcome**.

The selected features are all health-related measurements except the target column.

These features will be used to predict whether a patient has diabetes.


```python
x = df.drop("Outcome",axis=1)
y = df["Outcome"]
```

```python
x.head()
```

```python
y.head()
```

### Observation

X contains all input features that describe each patient.

y contains the Outcome values (0 = No Diabetes, 1 = Diabetes), which the model will learn to predict.

## Train/Test Split

The dataset is divided into training and testing sets.

80% of the data is used for training, while 20% is reserved for testing.

The parameter `stratify=y` ensures that both the training and testing sets keep approximately the same proportion of diabetic and non-diabetic patients.

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print("X train shape:", x_train.shape)
print("X test shape:", x_test.shape)
print("y train shape:", y_train.shape)
print("y test shape:", y_test.shape)
```

## Step 1: Train a LogisticRegression model on a provided classification dataset (e.g. churn).

A Logistic Regression model is created and trained using the training dataset.

The model learns the relationship between the input features and the target variable to classify whether a patient has diabetes.

```python
model = LogisticRegression(max_iter=1000)
model.fit(x_train,y_train)
print("model trained successfully")
```

### Observation

The Logistic Regression model was successfully trained using the training dataset.

During training, the model learned the relationship between the health-related features and the Outcome variable.

The trained model is now ready to classify new patients as diabetic or non-diabetic.

## Step 2: Generate predictions and produce the confusion matrix.

After training the model,predictions are generated for the test dataset

```python
pre = model.predict(x_test)
pre[:10] 
```

0 = NO diabets 
1 = Diabets 

```python
comp =pd.DataFrame({
    "actual" :y_test.values,
    "predected":pre
})
comp.head(10)
```

### Observation

The model generated predictions for the test dataset.

Each prediction is classified as either 0 (No Diabetes) or 1 (Diabetes).

Some predictions match the actual values, while others may differ, which is expected in real-world classification problems.

## Confusion Matrix

The confusion matrix shows how many predictions were correct and incorrect for each class.

```python
cm = confusion_matrix(y_test,pre)
print(cm)
```

[[TN FP]
 [FN TP]]

```python
cm_df = pd.DataFrame(
    cm,
    index=["Actual 0", "Actual 1"],
    columns=["Predicted 0", "Predicted 1"]
)

cm_df
```

### Observation

The confusion matrix shows that the model correctly classified **82 non-diabetic patients (True Negatives)** and **28 diabetic patients (True Positives)**.

However, the model incorrectly predicted **18 non-diabetic patients as diabetic (False Positives)** and missed **26 diabetic patients by predicting them as non-diabetic (False Negatives)**.

Overall, the model correctly classified more samples than it misclassified, but reducing the number of false negatives would improve its performance for diabetes prediction.

 ## Step 3: Compute precision, recall, and F1 with classification_report, and interpret each.

```python
print(classification_report(y_test,pre))
```

### Observation

The model works better for class 0 than class 1.

It correctly identifies most non-diabetic patients, but it misses some diabetic patients because the recall for class 1 is lower.

The recall for class 1 still needs improvement.

## STEP 4 : Decide whether precision or recall matters more for this specific problem and justify it.

For diabetes prediction, recall is more important than precision.
A higher recall helps the model detect more patients who actually have diabetes
Missing a diabetic patient is more serious than predicting a healthy patient as diabetic.

## Step 5: Compute the AUC-ROC and document what it says about the model.

```python
prob = model.predict_proba(x_test)[:,1]
auc = roc_auc_score(y_test,prob)
print("Auc , Roc ",auc )
```

The AUC-ROC score is 0.823, which is a good score.

This means the model can usually separate diabetic and non-diabetic patients.
