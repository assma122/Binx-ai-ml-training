<h2 style="color:#1D4ED8;margin:0 0 8px 0;">2. Why Do Dense Networks Struggle with Images?</h2>
<p style="color:#374151;line-height:1.8;margin:0;">Images contain far more inputs than ordinary tabular data. A dense layer connects <b>every input value to every neuron</b>, which can create millions of trainable parameters while ignoring spatial relationships between nearby pixels.</p>
</div>

<div style="background:#F5F3FF;border:1px solid #DDD6FE;border-radius:14px;padding:18px 20px;margin:15px 0;text-align:center;">
<p style="color:#374151;margin:0 0 8px 0;">A single RGB image with size</p>
<p style="color:#6D28D9;font-size:22px;font-weight:bold;margin:0;">200 × 200 × 3 = 120,000 pixel values</p>
<p style="color:#6B7280;margin:9px 0 0 0;">Width × Height × Color Channels</p>
</div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:10px;padding:12px 15px;margin:14px 0;">
<p style="color:#1E3A8A;line-height:1.7;margin:0;"><b>Why this code?</b> The training guide says a 200 × 200 RGB image contains 120,000 values. I will calculate the first dense layer's parameter count directly so the phrase “millions of weights” becomes a measurable result.</p>
</div>

```python
image_height = 200
image_width = 200
channels = 3
dense_neurons = 64

image_inputs = image_height * image_width * channels

dense_parameters = (
    image_inputs * dense_neurons
    + dense_neurons
)

print("Image input values:", f"{image_inputs:,}")
print("Dense layer parameters:", f"{dense_parameters:,}")
```

<div style="background:#FDF2F8;border-left:5px solid #EC4899;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#BE185D;">Reading the result:</b> A single <b style="color:#7C3AED;">200 × 200 RGB image</b> produces <b style="color:#2563EB;">120,000 inputs</b>. Connecting them to only <b>64 dense neurons</b> already requires <b style="color:#DB2777;">7,680,064 trainable parameters</b> in the first layer alone.</p>
</div>

<div style="background:#FFF7ED;border-left:5px solid #F97316;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#C2410C;">The second problem — spatial structure:</b> Flattening an image turns its pixels into one long sequence. The dense layer no longer directly uses the fact that neighboring pixels belong to nearby locations. For edges, shapes, and textures, that local relationship is important.</p>
</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#047857;">What we need instead:</b> A method that looks at <b style="color:#7C3AED;">small local regions</b>, preserves spatial relationships, and reuses the same pattern detector across the image. That is exactly what <b style="color:#059669;">convolution</b> provides.</p>
</div>

<div style="background:#F5F3FF;border-left:5px solid #8B5CF6;border-radius:12px;padding:16px 19px;margin:20px 0;">
<h2 style="color:#6D28D9;margin:0 0 8px 0;">3. Convolution — Looking at Images Locally</h2>
<p style="color:#374151;line-height:1.8;margin:0;">Instead of connecting every pixel to every neuron, convolution examines <b>small local regions</b>. A small matrix called a <b style="color:#7C3AED;">filter</b> or <b style="color:#DB2777;">kernel</b> slides across the image and responds strongly when it encounters the pattern it is designed to detect.</p>
</div>

<div style="background:#EFF6FF;border:1px solid #BFDBFE;border-radius:14px;padding:18px 20px;margin:16px 0;text-align:center;">
<p style="font-size:17px;font-weight:bold;margin:0;line-height:2.2;"><span style="color:#2563EB;">Image Region</span> → <span style="color:#7C3AED;">3 × 3 Filter</span> → <span style="color:#DB2777;">Multiply & Sum</span> → <span style="color:#059669;">Feature Map Value</span></p>
</div>

<div style="background:#FFF7ED;border-left:5px solid #F97316;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#C2410C;">How does one convolution step work?</b> The filter is placed over a small region. Each filter value is multiplied by the pixel underneath it, and the products are added together. The result becomes <b>one value</b> in the feature map.</p>
</div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:10px;padding:12px 15px;margin:14px 0;">
<p style="color:#1E3A8A;line-height:1.7;margin:0;"><b>Why this code?</b> Before applying convolution to a full image, I will calculate one small 3 × 3 example manually. This isolates the core operation and makes it clear how a filter turns one local image region into one feature-map value.</p>
</div>

```python
back = np.array([
    [10, 10, 80],
    [10, 10, 80],
    [10, 10, 80]
])

edge_filter = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
])

convolution_value = np.sum(back * edge_filter)

print("Image region:")
print(back)
print("\nEdge filter:")
print(edge_filter)
print("\nConvolution result:", convolution_value)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#047857;">Reading the calculation:</b> The left side of the region has lower values while the right side has higher values. The edge filter compares the two sides, producing a strong response where pixel intensity changes sharply.</p>
</div>

<div style="background:#FDF2F8;border-left:5px solid #EC4899;border-radius:12px;padding:16px 19px;margin:16px 0;">
<p style="color:#374151;line-height:1.9;margin:0;"><b style="color:#BE185D;">Filter / Kernel:</b> A small grid of weights that searches for a particular visual pattern, such as an edge or texture.<br><br><b style="color:#BE185D;">Feature Map:</b> The output created after the filter moves across the image. Strong values show locations where the filter responded to its target pattern.</p>
</div>

<div style="background:#FFFBEB;border-left:5px solid #F59E0B;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.9;margin:0;"><b style="color:#B45309;">Two controls affect how the filter moves:</b><br><br><b style="color:#7C3AED;">Stride</b> controls how many pixels the filter moves at each step. A stride of 1 moves one pixel at a time.<br><br><b style="color:#2563EB;">Padding</b> adds pixels around the border so the filter can process information near the image edges and, depending on the setting, preserve output size.</p>
</div>

<div style="background:linear-gradient(135deg,#EEF2FF 0%,#FDF2F8 100%);border:1px solid #DDD6FE;border-radius:15px;padding:18px 20px;margin:20px 0;">
<h3 style="color:#6D28D9;margin:0 0 8px 0;">Interactive Convolution Playground</h3>
<p style="color:#374151;line-height:1.8;margin:0;">
Before using the real dataset, I can experiment with several hand-defined filters.
Changing the filter updates the feature map immediately, making it easier to see
that different kernels respond to different visual patterns.
</p>
</div>


<div style="background:#F0F9FF;border-left:5px solid #0EA5E9;border-radius:11px;padding:13px 16px;margin:14px 0;">
<p style="color:#374151;line-height:1.8;margin:0;">
<b style="color:#0369A1;">Why this code? — Build an interactive filter explorer</b><br>
This cell creates a small synthetic image and lets me switch between <b>Vertical Edge</b>, <b>Horizontal Edge</b>, and <b>Sharpen</b> filters. The goal is to observe how the selected kernel changes the resulting feature map.
</p>
</div>


```python
import ipywidgets as widgets
from IPython.display import display, clear_output

demo_image = np.array([
    [20, 20, 20, 20, 90, 90, 90, 90],
    [20, 20, 20, 20, 90, 90, 90, 90],
    [20, 20, 70, 70, 120, 120, 90, 90],
    [20, 20, 70, 70, 120, 120, 90, 90],
    [20, 20, 70, 70, 120, 120, 90, 90],
    [20, 20, 70, 70, 120, 120, 90, 90],
    [20, 20, 20, 20, 90, 90, 90, 90],
    [20, 20, 20, 20, 90, 90, 90, 90]
], dtype=float)

filters = {
    "Vertical Edge": np.array([[-1, 0, 1],
                               [-1, 0, 1],
                               [-1, 0, 1]]),
    "Horizontal Edge": np.array([[-1, -1, -1],
                                 [ 0,  0,  0],
                                 [ 1,  1,  1]]),
    "Sharpen": np.array([[ 0, -1,  0],
                         [-1,  5, -1],
                         [ 0, -1,  0]])
}

def manual_convolution(image, kernel, stride=1, padding=0):
    if padding > 0:
        image = np.pad(image, padding, mode="constant")
    kh, kw = kernel.shape
    out_h = (image.shape[0] - kh) // stride + 1
    out_w = (image.shape[1] - kw) // stride + 1
    output = np.zeros((out_h, out_w))

    for r in range(out_h):
        for c in range(out_w):
            region = image[
                r * stride:r * stride + kh,
                c * stride:c * stride + kw
            ]
            output[r, c] = np.sum(region * kernel)
    return output

filter_dropdown = widgets.Dropdown(
    options=list(filters.keys()),
    value="Vertical Edge",
    description="Filter:"
)

output_box = widgets.Output()

def update_filter(change=None):
    with output_box:
        clear_output(wait=True)
        selected = filters[filter_dropdown.value]
        result = manual_convolution(demo_image, selected)

        fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))
        axes[0].imshow(demo_image, cmap="gray")
        axes[0].set_title("Input")
        axes[0].axis("off")

        axes[1].imshow(selected, cmap="coolwarm")
        axes[1].set_title(filter_dropdown.value)
        axes[1].set_xticks([])
        axes[1].set_yticks([])

        axes[2].imshow(result, cmap="gray")
        axes[2].set_title("Feature Map")
        axes[2].axis("off")

        plt.tight_layout()
        plt.show()

filter_dropdown.observe(update_filter, names="value")
display(filter_dropdown, output_box)
update_filter()
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;">
<b style="color:#047857;">What to explore:</b>
The vertical-edge filter responds strongly to left-to-right intensity changes,
the horizontal-edge filter responds to top-to-bottom changes, and the sharpen
filter emphasizes local contrast. The input stays the same — only the kernel changes.
</p>
</div>


<div style="background:#FFFBEB;border-left:5px solid #F59E0B;border-radius:12px;padding:16px 19px;margin:20px 0;">
<h3 style="color:#B45309;margin:0 0 8px 0;">Interactive Stride & Padding Lab</h3>
<p style="color:#374151;line-height:1.8;margin:0;">
Stride changes how far the filter moves, while padding changes how the image borders
are handled. The controls below make their effect on the output size visible.
</p>
</div>


<div style="background:#F0F9FF;border-left:5px solid #0EA5E9;border-radius:11px;padding:13px 16px;margin:14px 0;">
<p style="color:#374151;line-height:1.8;margin:0;">
<b style="color:#0369A1;">Why this code? — See stride and padding instead of memorizing them</b><br>
This cell lets me change <b>stride</b> and <b>padding</b> and immediately compare the resulting feature map shape. It turns the output-size rule into a visible experiment.
</p>
</div>


```python
stride_slider = widgets.IntSlider(
    value=1, min=1, max=3, step=1, description="Stride:"
)
padding_toggle = widgets.ToggleButtons(
    options=[("No Padding", 0), ("Padding = 1", 1)],
    value=0,
    description="Padding:"
)

stride_output = widgets.Output()

def update_stride_padding(change=None):
    with stride_output:
        clear_output(wait=True)

        kernel = filters["Vertical Edge"]
        result = manual_convolution(
            demo_image,
            kernel,
            stride=stride_slider.value,
            padding=padding_toggle.value
        )

        print("Input shape :", demo_image.shape)
        print("Stride      :", stride_slider.value)
        print("Padding     :", padding_toggle.value)
        print("Output shape:", result.shape)

        plt.figure(figsize=(5, 3.5))
        plt.imshow(result, cmap="gray")
        plt.title(
            f"Feature Map | stride={stride_slider.value}, "
            f"padding={padding_toggle.value}"
        )
        plt.axis("off")
        plt.show()

stride_slider.observe(update_stride_padding, names="value")
padding_toggle.observe(update_stride_padding, names="value")

display(stride_slider, padding_toggle, stride_output)
update_stride_padding()
```

<p style="color:#374151;line-height:1.8;margin:0;">
<b style="color:#BE185D;">Experiment takeaway:</b>
A larger stride samples fewer positions and therefore produces a smaller feature map.
Padding adds a border around the input, allowing the filter to process locations near
the original image boundary and potentially preserve more spatial size.
</p>
</div>

I also resize it to 256 × 256 pixels to make the manual convolution fast and consistent while keeping enough visual detail for the demonstration.</p>
</div>

```python
image = Image.open(image_path).convert("L").resize((256, 256))
image_array = np.array(image, dtype=np.float32)

print("Image shape used for convolution:", image_array.shape)

plt.figure(figsize=(5, 5))
plt.imshow(image_array, cmap="gray")
plt.title("Selected Grayscale Image")
plt.axis("off")
plt.show()
```

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:10px;padding:12px 15px;margin:14px 0;">
<p style="color:#1E3A8A;line-height:1.7;margin:0;"><b>Why this code?</b> This cell applies the same 3 × 3 vertical edge filter to every valid image location. The nested loops deliberately show the sliding-window logic instead of hiding it inside a deep-learning framework.</p>
</div>

```python
edge_filter = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
], dtype=np.float32)

height, width = image_array.shape
feature_map = np.zeros((height - 2, width - 2), dtype=np.float32)

for row in range(height - 2):
    for col in range(width - 2):
        image_region = image_array[row:row + 3, col:col + 3]
        feature_map[row, col] = np.sum(image_region * edge_filter)
```

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:10px;padding:12px 15px;margin:14px 0;">
<p style="color:#1E3A8A;line-height:1.7;margin:0;"><b>Why this code?</b> A convolution is easier to understand visually than as a matrix of numbers. I will place the original image, the hand-defined kernel, and the resulting feature map side by side so the transformation can be inspected directly.</p>
</div>

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].imshow(image_array, cmap="gray")
axes[0].set_title("Original Image")
axes[0].axis("off")

axes[1].imshow(edge_filter, cmap="coolwarm")
axes[1].set_title("3 × 3 Edge Filter")
axes[1].set_xticks([])
axes[1].set_yticks([])

axes[2].imshow(feature_map, cmap="gray")
axes[2].set_title("Resulting Feature Map")
axes[2].axis("off")

plt.suptitle(
    "Original Image → Filter → Feature Map",
    fontsize=15,
    fontweight="bold"
)
plt.tight_layout()
plt.show()
```

<div style="background:#FDF2F8;border-left:5px solid #EC4899;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#BE185D;">Reading the feature map:</b> The original image contains a darker lesion region surrounded by lighter skin. After convolution, the output emphasizes <b>changes in intensity</b> rather than preserving the original appearance. Boundaries and texture variations become more visible, showing how a filter can extract structural information.</p>
</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#047857;">What this demonstrates:</b> The same small <b style="color:#7C3AED;">3 × 3 filter</b> was reused across the entire image. Convolution searches for the same visual pattern wherever it appears while keeping information about <b>where</b> the response occurred.</p>
</div>

<div style="background:#FFFBEB;border-left:5px solid #F59E0B;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#B45309;">Important distinction:</b> This filter is hand-defined only to make convolution visible. In a real CNN, useful filter values are <b>learned automatically during training</b>, allowing different filters to respond to edges, textures, shapes, and richer visual patterns.</p>
</div>

<h2 style="color:#047857;margin:0 0 8px 0;">5. Hands-On Step 3 — Parameter Sharing & CNN Efficiency</h2>
<p style="color:#374151;line-height:1.8;margin:0;">The previous experiment used the <b>same small filter</b> at every location. This gives CNNs two important advantages: <b style="color:#7C3AED;">parameter sharing</b> and <b style="color:#DB2777;">translation invariance</b>.</p>
</div>

<div style="background:#F5F3FF;border-left:5px solid #8B5CF6;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#6D28D9;">Parameter Sharing:</b> A convolutional filter does not need a different set of weights for every location. The <b>same weights are reused</b> as the filter slides across the entire image.</p>
</div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:10px;padding:12px 15px;margin:14px 0;">
<p style="color:#1E3A8A;line-height:1.7;margin:0;"><b>Why this code?</b> I will compare an example dense first layer with a small Conv2D layer using exact parameter counts. This makes the effect of parameter sharing visible instead of describing it only in words.</p>
</div>

```python
image_inputs = 200 * 200 * 3
dense_neurons = 64

dense_parameters = image_inputs * dense_neurons + dense_neurons

conv_filters = 32
kernel_height = 3
kernel_width = 3
input_channels = 3

conv_parameters = (
    kernel_height * kernel_width * input_channels + 1
) * conv_filters

parameter_comparison = pd.DataFrame({
    "Layer": ["Dense Layer", "Conv2D Layer"],
    "Configuration": [
        "120,000 inputs → 64 neurons",
        "32 filters of size 3 × 3 × 3"
    ],
    "Trainable Parameters": [dense_parameters, conv_parameters]
})

parameter_comparison
```

<div style="background:#FDF2F8;border-left:5px solid #EC4899;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#BE185D;">Reading the comparison:</b> The dense layer requires <b style="color:#DB2777;">7,680,064 parameters</b>, while the example Conv2D layer requires only <b style="color:#059669;">896 parameters</b>. The reduction comes from reusing each filter across the image instead of creating separate connections for every pixel location.</p>
</div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:10px;padding:12px 15px;margin:14px 0;">
<p style="color:#1E3A8A;line-height:1.7;margin:0;"><b>Why this code?</b> The table gives the exact numbers, but a visual comparison makes the scale difference easier to notice immediately. A logarithmic y-axis keeps both the 896-parameter Conv2D layer and the multi-million-parameter dense layer visible in the same chart.</p>
</div>

```python
plt.figure(figsize=(7, 4.5))
plt.bar(
    parameter_comparison["Layer"],
    parameter_comparison["Trainable Parameters"],
    color=["#EC4899", "#10B981"]
)
plt.yscale("log")
plt.ylabel("Trainable Parameters (log scale)")
plt.title("Dense vs Conv2D Parameter Count")
plt.grid(axis="y", alpha=0.2)
plt.show()
```

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#1D4ED8;">Translation Invariance:</b> Because the same filter scans different locations, a learned pattern can still be detected when it appears in another part of the image. The network does not need a completely different detector simply because the pattern moved.</p>
</div>

<div style="background:#FFF7ED;border-left:5px solid #F97316;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#C2410C;">Connection to the experiment:</b> The edge filter scanned the complete image rather than one fixed location. Similar intensity changes could therefore produce responses wherever they appeared, allowing one small filter to create an entire feature map.</p>
</div>

<div style="background:#F5F3FF;border:1px solid #DDD6FE;border-radius:14px;padding:18px 20px;margin:16px 0;text-align:center;">
<p style="color:#6D28D9;font-size:17px;font-weight:bold;margin:0;line-height:2.2;"><span style="color:#2563EB;">Edges</span> → <span style="color:#DB2777;">Textures & Shapes</span> → <span style="color:#059669;">Higher-Level Visual Patterns</span></p>
</div>

<div style="background:#FFFBEB;border-left:5px solid #F59E0B;border-radius:11px;padding:14px 17px;margin:15px 0;">
<p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#B45309;">Feature hierarchy:</b> Early CNN layers tend to respond to simple patterns such as edges. Middle layers combine those responses into textures and shapes, while deeper layers can represent higher-level structures. These features are learned from data rather than manually designed.</p>
</div>

<b style="color:#0369A1;">Why this code? — Check my understanding</b><br>
This is not a model-training step. It is a quick self-check that gives immediate feedback on the main ideas from today's lesson.

```python
quiz_questions = [
    {
        "question": "What happens when stride increases?",
        "options": [
            "The filter moves in larger steps",
            "The image gains more color channels",
            "The kernel automatically becomes larger"
        ],
        "answer": "The filter moves in larger steps"
    },
    {
        "question": "Why does parameter sharing reduce CNN parameters?",
        "options": [
            "The same filter weights are reused across image locations",
            "Every pixel gets its own filter",
            "CNNs remove all image pixels"
        ],
        "answer": "The same filter weights are reused across image locations"
    },
    {
        "question": "What does a feature map show?",
        "options": [
            "Where a filter responds to its pattern",
            "Only the original RGB values",
            "The number of classes in the dataset"
        ],
        "answer": "Where a filter responds to its pattern"
    }
]

question_selector = widgets.Dropdown(
    options=[f"Question {i+1}" for i in range(len(quiz_questions))],
    value="Question 1",
    description="Choose:",
    layout=widgets.Layout(width="380px")
)

question_text = widgets.HTML()

answer_selector = widgets.RadioButtons(
    layout=widgets.Layout(
        width="650px",
        min_height="110px"
    )
)

check_button = widgets.Button(
    description="Check Answer",
    button_style="info",
    layout=widgets.Layout(width="190px")
)

feedback = widgets.HTML()


def load_question(change=None):
    q_index = int(question_selector.value.split()[-1]) - 1
    q = quiz_questions[q_index]

    # Clear feedback from the previous question
    feedback.value = ""

    # Update question and answers
    question_text.value = (
        f"<b style='color:#6D28D9;font-size:15px;'>"
        f"{q['question']}</b>"
    )

    answer_selector.options = q["options"]
    answer_selector.value = None


def check_answer(_):
    q_index = int(question_selector.value.split()[-1]) - 1
    q = quiz_questions[q_index]

    if answer_selector.value is None:
        feedback.value = (
            "<b style='color:#B45309;'>"
            "Choose an answer first.</b>"
        )

    elif answer_selector.value == q["answer"]:
        feedback.value = (
            "<b style='color:#047857;'>"
            "Correct ✓</b>"
        )

    else:
        feedback.value = (
            "<b style='color:#BE185D;'>"
            "Try again.</b> Revisit the visual explanation above."
        )


question_selector.observe(load_question, names="value")
check_button.on_click(check_answer)

display(
    question_selector,
    question_text,
    answer_selector,
    check_button,
    feedback
)

load_question()
```
