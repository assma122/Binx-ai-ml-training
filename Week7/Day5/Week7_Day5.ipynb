<div style="color:#475569;font-size:13.5px;line-height:1.72;margin-top:8px;">Import the tools used for data handling, visualization, leakage-safe preprocessing, Gradient Boosting, cross-validation, threshold selection, and evaluation. Imports are kept in one place to avoid duplication.</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, HTML
from matplotlib.colors import LinearSegmentedColormap

from sklearn.model_selection import (
    train_test_split,
    GridSearchCV,
    StratifiedKFold,
    cross_val_predict,
    learning_curve
)
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
from sklearn.ensemble import HistGradientBoostingClassifier

# Consistent presentation palette used throughout the notebook.
PURPLE = "#7C3AED"
PINK = "#EC4899"
BLUE = "#2563EB"
TEAL = "#0891B2"
ORANGE = "#F97316"
GREEN = "#16A34A"
GOLD = "#D4A017"
RED = "#DC2626"
NAVY = "#172554"
LIGHT_GRID = "#CBD5E1"
```

<span style="color:#1D4ED8;font-size:17px;font-weight:900;">Load and Inspect the Project Dataset</span>

<div style="color:#475569;font-size:13.5px;line-height:1.72;margin-top:8px;">Load the same Heart Failure Prediction dataset used in Sprint 1 so that every model comparison remains tied to the same prediction problem and split strategy.</div>

```python
data_path = "heart.csv"
df = pd.read_csv(data_path)

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
print("\nMissing values before clinical cleaning:")
print(df.isna().sum())
display(df.head())
```

<span style="background:#2563EB;color:white;border-radius:8px;padding:4px 8px;font-size:12px;font-weight:900;">2.4</span>
<span style="color:#1D4ED8;font-size:17px;font-weight:900;">Prepare the Training and Testing Data</span>
</div>
<div style="color:#475569;font-size:13.5px;line-height:1.72;margin-top:8px;">Apply the same clinical cleaning rule used previously: zero values in <code>Cholesterol</code> and <code>RestingBP</code> are treated as missing. Then create the original 80/20 stratified train-test split. The test set is reserved and is not used during model selection.</div>
</div>

```python
df_clean = df.copy()
df_clean[["Cholesterol", "RestingBP"]] = (
    df_clean[["Cholesterol", "RestingBP"]].replace(0, np.nan)
)

X = df_clean.drop(columns="HeartDisease")
y = df_clean["HeartDisease"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

split_summary = pd.DataFrame({
    "Split": ["Training pool", "Reserved test"],
    "Rows": [len(X_train), len(X_test)],
    "Positive Rate": [y_train.mean(), y_test.mean()]
})

display(
    split_summary.style
    .format({"Positive Rate": "{:.1%}"})
    .set_properties(**{"text-align": "center"})
    .set_table_styles([
        {"selector":"th","props":[("background-color", NAVY),("color","white"),("font-weight","bold"),("text-align","center")]},
        {"selector":"td","props":[("padding","7px 10px")]}
    ])
)
```

<div style="background:#EFF6FF;border:1px solid #2563EB33;border-radius:14px;padding:14px 17px;margin:17px 0 9px 0;">
<div style="display:flex;align-items:center;gap:10px;">
<span style="background:#2563EB;color:white;border-radius:8px;padding:4px 8px;font-size:12px;font-weight:900;">2.5</span>
<span style="color:#1D4ED8;font-size:17px;font-weight:900;">Build Leakage-Safe Preprocessing Inside a Pipeline</span>
</div>
<div style="color:#475569;font-size:13.5px;line-height:1.72;margin-top:8px;">Median imputation is applied to numerical features and One-Hot Encoding to categorical features. Keeping preprocessing inside a <code>ColumnTransformer</code> and <code>Pipeline</code> ensures that transformations are learned only from the training portion available inside each fit or cross-validation fold.</div>
</div>

```python
numerical_features = [
    "Age", "RestingBP", "Cholesterol", "MaxHR", "Oldpeak"
]

categorical_features = [
    "Sex", "ChestPainType", "FastingBS",
    "RestingECG", "ExerciseAngina", "ST_Slope"
]

numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline([
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer([
    ("numerical", numerical_pipeline, numerical_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocessing pipeline ready.")
print("Numerical features:", len(numerical_features))
print("Categorical features:", len(categorical_features))
```

<div style="background:#EFF6FF;border:1px solid #2563EB33;border-radius:14px;padding:14px 17px;margin:17px 0 9px 0;">
<div style="display:flex;align-items:center;gap:10px;">
<span style="background:#2563EB;color:white;border-radius:8px;padding:4px 8px;font-size:12px;font-weight:900;">2.6</span>
<span style="color:#1D4ED8;font-size:17px;font-weight:900;">Create a Validation Set for Initial and Manual Experiments</span>
</div>
<div style="color:#475569;font-size:13.5px;line-height:1.72;margin-top:8px;">Split only the training pool to create a validation set. Initial and manual Gradient Boosting experiments are evaluated here, while the final test set remains untouched.</div>
</div>

```python
X_gb_train, X_val, y_gb_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=42
)

print("Manual-training rows:", len(X_gb_train))
print("Validation rows:", len(X_val))
print("Reserved test rows:", len(X_test))
```

<div style="background:#FFF7ED;border:1px solid #F9731633;border-radius:14px;padding:14px 17px;margin:17px 0 9px 0;">
<div style="display:flex;align-items:center;gap:10px;">
<span style="background:#F97316;color:white;border-radius:8px;padding:4px 8px;font-size:12px;font-weight:900;">2.7</span>
<span style="color:#C2410C;font-size:17px;font-weight:900;">Experiment 1 — Initial Gradient Boosting</span>
</div>
<div style="color:#475569;font-size:13.5px;line-height:1.72;margin-top:8px;">Train a default <code>HistGradientBoostingClassifier</code> on the manual-training split and evaluate it only on the validation set. This becomes the Sprint 2 starting point.</div>
</div>

```python
initial_gb_model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", HistGradientBoostingClassifier(random_state=42))
])

initial_gb_model.fit(X_gb_train, y_gb_train)
initial_val_predictions = initial_gb_model.predict(X_val)

initial_val_metrics = {
    "Accuracy": accuracy_score(y_val, initial_val_predictions),
    "Precision": precision_score(y_val, initial_val_predictions),
    "Recall": recall_score(y_val, initial_val_predictions),
    "F1-score": f1_score(y_val, initial_val_predictions)
}

initial_val_results = pd.DataFrame(
    [initial_val_metrics],
    index=["Initial Gradient Boosting"]
)

display(
    initial_val_results.style
    .format("{:.3f}")
)

```

<div style="font-size:13.5px;color:#475569;line-height:1.7;">
The initial Gradient Boosting model provides a solid starting point, achieving an F1-score of approximately <b>0.835</b> and Recall of approximately <b>0.877</b> on the validation set. These results establish the Sprint 2 starting point for evaluating whether the following tuning experiments can improve model performance.
</div>

</div>

<div style="background:#FFF7ED;border:1px solid #F9731633;border-radius:14px;padding:14px 17px;margin:17px 0 9px 0;">
<div style="display:flex;align-items:center;gap:10px;">
<span style="background:#F97316;color:white;border-radius:8px;padding:4px 8px;font-size:12px;font-weight:900;">2.8</span>
<span style="color:#C2410C;font-size:17px;font-weight:900;">Define Four Manual Tuning Experiments</span>
</div>
<div style="color:#475569;font-size:13.5px;line-height:1.72;margin-top:8px;">Change learning rate, boosting iterations, leaf capacity, and L2 regularization. Every configuration is explicit so the experiment history is reproducible.</div>
</div>

```python
experiment_configs = [
    {"name":"GB_01", "learning_rate":0.05, "max_iter":100, "max_leaf_nodes":15, "l2_regularization":0.0},
    {"name":"GB_02", "learning_rate":0.05, "max_iter":200, "max_leaf_nodes":15, "l2_regularization":0.1},
    {"name":"GB_03", "learning_rate":0.10, "max_iter":100, "max_leaf_nodes":31, "l2_regularization":0.1},
    {"name":"GB_04", "learning_rate":0.03, "max_iter":300, "max_leaf_nodes":15, "l2_regularization":1.0}
]

pd.DataFrame(experiment_configs)
```

<div style="background:#FFF7ED;border:1px solid #F9731633;border-radius:14px;padding:14px 17px;margin:17px 0 9px 0;">
<div style="display:flex;align-items:center;gap:10px;">
<span style="background:#F97316;color:white;border-radius:8px;padding:4px 8px;font-size:12px;font-weight:900;">2.9</span>
<span style="color:#C2410C;font-size:17px;font-weight:900;">Run, Log, and Style the Manual Experiments</span>
</div>
<div style="color:#475569;font-size:13.5px;line-height:1.72;margin-top:8px;">Fit each configuration on <code>X_gb_train</code>, evaluate it on <code>X_val</code>, and log both the configuration and the validation metrics. The best manual experiment is highlighted, but it is not automatically treated as the final model.</div>
</div>

```python
experiment_results = []
trained_experiments = {}

for config in experiment_configs:
    model = Pipeline([
        ("preprocessing", preprocessor),
        ("classifier", HistGradientBoostingClassifier(
            learning_rate=config["learning_rate"],
            max_iter=config["max_iter"],
            max_leaf_nodes=config["max_leaf_nodes"],
            l2_regularization=config["l2_regularization"],
            random_state=42
        ))
    ])

    model.fit(X_gb_train, y_gb_train)
    val_predictions = model.predict(X_val)

    experiment_results.append({
        "Experiment": config["name"],
        "Learning Rate": config["learning_rate"],
        "Max Iterations": config["max_iter"],
        "Max Leaf Nodes": config["max_leaf_nodes"],
        "L2 Regularization": config["l2_regularization"],
        "Validation Accuracy": accuracy_score(y_val, val_predictions),
        "Validation Precision": precision_score(y_val, val_predictions),
        "Validation Recall": recall_score(y_val, val_predictions),
        "Validation F1": f1_score(y_val, val_predictions)
    })
    trained_experiments[config["name"]] = model

experiment_results_df = pd.DataFrame(experiment_results)
best_manual_row = experiment_results_df.loc[
    experiment_results_df["Validation F1"].idxmax()
]

display(
    experiment_results_df.style
    .format({
        "Learning Rate":"{:.2f}",
        "L2 Regularization":"{:.1f}",
        "Validation Accuracy":"{:.3f}",
        "Validation Precision":"{:.3f}",
        "Validation Recall":"{:.3f}",
        "Validation F1":"{:.3f}"
    }))

```

</div>
<div style="color:#475569;font-size:13.5px;line-height:1.72;margin-top:8px;">A single validation split can be sensitive to which records happen to fall into it. GridSearchCV searches a broader parameter space and evaluates each candidate across five stratified folds using F1-score.</div>
</div>

```python
cv_strategy = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

advanced_pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", HistGradientBoostingClassifier(random_state=42))
])

param_grid = {
    "classifier__learning_rate": [0.02, 0.03, 0.05, 0.08],
    "classifier__max_iter": [150, 250, 350],
    "classifier__max_leaf_nodes": [7, 15, 31],
    "classifier__min_samples_leaf": [10, 20, 30],
    "classifier__l2_regularization": [0.0, 0.5, 1.0]
}

grid_search = GridSearchCV(
    estimator=advanced_pipeline,
    param_grid=param_grid,
    scoring="f1",
    cv=cv_strategy,
    n_jobs=-1,
    verbose=1,
    return_train_score=True
)

total_combinations = int(np.prod([len(values) for values in param_grid.values()]))
print("Parameter combinations:", total_combinations)
print("Total CV fits:", total_combinations * cv_strategy.get_n_splits())

grid_search.fit(X_train, y_train)

print("\nBest parameters:")
print(grid_search.best_params_)
print("\nBest mean CV F1-score:", round(grid_search.best_score_, 3))
```
