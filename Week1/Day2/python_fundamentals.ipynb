1) This function returns the mean,max,min values of a list 

```python
def caculate_num(numbers):
   return {
    "mean":sum(numbers)/len(numbers),
    "min":min(numbers),
    "max":max(numbers)

   }
result = caculate_num([10,20,30,40,50])
print(result)
```

2) this code filters the even numbers using list comprehension

```python
numbers = range(21)
even=[k for k in numbers if k%2==0]
print(even)
```

3) small class : this class has two attributes and one method to show the book information

```python
class Book:
    def __init__(self,title,pages):
        self.title = title 
        self.pages= pages 

    def display(self):
        return f"{self.title} has {self.pages} pages "    
book = Book("python",130)

print(book.title)
print(book.pages)
print(book.display())
```
