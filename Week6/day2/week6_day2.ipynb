  <h2 style="color:#BE185D;margin:0 0 7px 0;">2. Common Activation Functions</h2>
  <p style="color:#374151;line-height:1.8;margin:0;">The appropriate activation depends on the layer and the prediction task.</p>
</div>

<div style="background:#F5F3FF;border:1px solid #DDD6FE;border-radius:12px;padding:15px 18px;margin:12px 0;">
  <h3 style="color:#6D28D9;margin:0 0 7px 0;">ReLU — Hidden Layers</h3>
  <p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#7C3AED;">Formula:</b> <code>ReLU(z) = max(0, z)</code><br>Negative values become zero, while positive values remain unchanged: <code>ReLU(-3) = 0</code> and <code>ReLU(5) = 5</code>.</p>
</div>

<div style="background:#EFF6FF;border:1px solid #BFDBFE;border-radius:12px;padding:15px 18px;margin:12px 0;">
  <h3 style="color:#1D4ED8;margin:0 0 7px 0;">Sigmoid — Binary Classification Output</h3>
  <p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#2563EB;">Formula:</b> <code>sigmoid(z) = 1 / (1 + e<sup>-z</sup>)</code><br>Sigmoid maps any input into the range <code>0</code> to <code>1</code>, making it suitable for one output neuron that represents an estimated probability.</p>
</div>

<div style="background:#ECFDF5;border:1px solid #A7F3D0;border-radius:12px;padding:15px 18px;margin:12px 0;">
  <h3 style="color:#047857;margin:0 0 7px 0;">Tanh — Zero-Centered Activation</h3>
  <p style="color:#374151;line-height:1.8;margin:0;">Tanh maps inputs into the range <code>-1</code> to <code>1</code>. Unlike Sigmoid, its outputs can be negative or positive, so they are centered around zero.</p>
</div>

<div style="background:#FFF7ED;border:1px solid #FED7AA;border-radius:12px;padding:15px 18px;margin:12px 0;">
  <h3 style="color:#C2410C;margin:0 0 7px 0;">Softmax — Multi-Class Classification</h3>
  <p style="color:#374151;line-height:1.8;margin:0;">Softmax converts multiple output scores into class probabilities that add up to <code>1</code>. It is useful for multi-class tasks, while a single Sigmoid output is the simpler choice for this binary heart-disease project.</p>
</div>


<div style="background:#F5F3FF;border:1px solid #DDD6FE;border-radius:13px;padding:16px 19px;margin:16px 0;">
  <h3 style="color:#6D28D9;margin:0 0 8px 0;">Why Not Use Sigmoid in Every Layer?</h3>
  <p style="color:#374151;line-height:1.8;margin:0;">Sigmoid is useful in the output layer because it produces a probability. However, when its outputs move close to <code>0</code> or <code>1</code>, its learning signal can become very small. Across multiple hidden layers, this can contribute to the <b style="color:#BE185D;">vanishing gradient problem</b>.</p>
  <p style="color:#374151;line-height:1.8;margin:9px 0 0 0;"><b style="color:#059669;">ReLU</b> is commonly preferred in hidden layers because it is simple and preserves stronger learning signals for positive inputs.</p>
</div>

<div style="background:#FFF7ED;border-left:5px solid #F97316;border-radius:10px;padding:12px 15px;margin:12px 0;">
  <p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#C2410C;">Limitation:</b> A neuron that repeatedly receives negative inputs may remain inactive because ReLU returns zero. This is sometimes called the <b style="color:#BE185D;">dying ReLU problem</b>.</p>
</div>


<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:13px;padding:16px 19px;margin:21px 0 13px 0;">
  <h2 style="color:#1D4ED8;margin:0 0 7px 0;">3. Forward Propagation</h2>
  <p style="color:#374151;line-height:1.8;margin:0;">Forward propagation moves patient features through the network, layer by layer, until the output neuron produces a prediction.</p>
</div>

<div style="background:#FAF5FF;border:1px solid #E9D5FF;border-radius:13px;padding:16px 19px;margin:14px 0;">
  <h3 style="color:#6D28D9;margin:0 0 8px 0;">The Proposed Network</h3>
  <p style="color:#374151;line-height:1.9;margin:0;">
    <b style="color:#2563EB;">Input layer:</b> 21 processed patient features.<br>
    <b style="color:#7C3AED;">First hidden layer:</b> 16 neurons using ReLU.<br>
    <b style="color:#DB2777;">Second hidden layer:</b> 8 neurons using ReLU.<br>
    <b style="color:#059669;">Output layer:</b> 1 neuron using Sigmoid.
  </p>
</div>

<div style="background:#FFF7ED;border:1px solid #FED7AA;border-radius:13px;padding:16px 19px;margin:14px 0;">
  <h3 style="color:#C2410C;margin:0 0 8px 0;">The Forward Pass in Three Transformations</h3>
  <p style="color:#374151;line-height:1.9;margin:0;">
    <b style="color:#7C3AED;">Hidden layer 1:</b> <code>h1 = ReLU(X · W1 + b1)</code><br>
    <b style="color:#DB2777;">Hidden layer 2:</b> <code>h2 = ReLU(h1 · W2 + b2)</code><br>
    <b style="color:#059669;">Output layer:</b> <code>prediction = sigmoid(h2 · W3 + b3)</code>
  </p>
</div>

<div style="background:#FEF2F2;border:1px solid #FECACA;border-radius:10px;padding:12px 15px;margin:12px 0;">
  <p style="color:#7F1D1D;line-height:1.8;margin:0;"><b>Important:</b> Forward propagation calculates a prediction using the current weights and biases. It does not update those parameters by itself.</p>
</div>


<div style="background:#FFF1F2;border-left:5px solid #F43F5E;border-radius:13px;padding:16px 19px;margin:21px 0 13px 0;">
  <h2 style="color:#BE123C;margin:0 0 7px 0;">4. Understanding the Loss Function</h2>
  <p style="color:#374151;line-height:1.8;margin:0;">A loss function compares a prediction with the actual label and measures how wrong the prediction is.</p>
</div>

<div style="background:#F8FAFC;border:1px solid #E2E8F0;border-radius:12px;padding:15px 18px;margin:13px 0;">
  <p style="color:#374151;line-height:1.9;margin:0;">
    <b style="color:#2563EB;">Prediction:</b> The probability produced by the model.<br>
    <b style="color:#DB2777;">True label:</b> The actual outcome in the dataset.<br>
    <b style="color:#EA580C;">Loss:</b> A numerical measure of prediction error.
  </p>
</div>

<div style="background:#FFF7ED;border:1px solid #FED7AA;border-radius:12px;padding:15px 18px;margin:13px 0;">
  <h3 style="color:#C2410C;margin:0 0 8px 0;">Binary Cross-Entropy</h3>
  <p style="color:#9A3412;font-weight:bold;line-height:1.8;margin:0 0 7px 0;"><code>Loss = -[y × ln(p) + (1 - y) × ln(1 - p)]</code></p>
  <p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#7C3AED;">y</b> is the actual class, either 0 or 1, and <b style="color:#DB2777;">p</b> is the predicted probability of the positive class.</p>
</div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:10px;padding:12px 15px;margin:12px 0;">
  <p style="color:#1E3A8A;line-height:1.8;margin:0;"><b>Accuracy vs. loss:</b> Accuracy considers the final predicted class, while Binary Cross-Entropy also considers how confident the predicted probability is.</p>
</div>

  <h2 style="color:#FFFFFF;margin:0 0 7px 0;">Visualizing Activation Functions</h2>
  <p style="color:#F5F3FF;line-height:1.7;margin:0;">I will define ReLU, Sigmoid, and Tanh, then plot them over the same input range.</p>
</div>


```python
import numpy as np
import matplotlib.pyplot as plt


def relu(x):
    return np.maximum(0, x)


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def tanh(x):
    return np.tanh(x)


x_values = np.linspace(-6, 6, 300)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].plot(x_values, relu(x_values), color="#7C3AED", linewidth=3)
axes[0].set_title("ReLU")
axes[0].set_xlabel("Input")
axes[0].set_ylabel("Output")
axes[0].grid(alpha=0.2)

axes[1].plot(x_values, sigmoid(x_values), color="#F43F5E", linewidth=3)
axes[1].set_title("Sigmoid")
axes[1].set_xlabel("Input")
axes[1].set_ylabel("Output")
axes[1].grid(alpha=0.2)

axes[2].plot(x_values, tanh(x_values), color="#059669", linewidth=3)
axes[2].set_title("Tanh")
axes[2].set_xlabel("Input")
axes[2].set_ylabel("Output")
axes[2].grid(alpha=0.2)

plt.suptitle("How Activation Functions Transform the Same Inputs", fontsize=14)
plt.tight_layout()
plt.show()
```

<div style="background:#EFF6FF;border:1px solid #BFDBFE;border-radius:13px;padding:16px 19px;margin:16px 0;">
  <h3 style="color:#1D4ED8;margin:0 0 8px 0;">Understanding the Proposed Architecture</h3>
  <p style="color:#374151;line-height:1.8;margin:0;">The proposed network contains <b style="color:#2563EB;">21 input features</b>, two hidden layers with <b style="color:#7C3AED;">16</b> and <b style="color:#DB2777;">8 neurons</b>, and <b style="color:#059669;">one output neuron</b>.</p>

<div style="background:#FFFFFF;border:1px solid #E5E7EB;border-radius:12px;padding:15px 18px;margin:13px 0;">

  <p style="color:#374151;line-height:1.9;margin:0;">
    <b style="color:#2563EB;">First hidden layer:</b> 21 × 16 weights + 16 biases = 352 parameters.<br>
    <b style="color:#DB2777;">Second hidden layer:</b> 16 × 8 weights + 8 biases = 136 parameters.<br>
    <b style="color:#059669;">Output layer:</b> 8 × 1 weights + 1 bias = 9 parameters.<br>
    <b style="color:#7C3AED;">Total:</b> 352 + 136 + 9 = <b style="color:#BE185D;">497 trainable parameters.</b>

<div style="background:#F5F3FF;border-left:5px solid #8B5CF6;border-radius:10px;padding:12px 15px;margin:12px 0;">
  <p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#6D28D9;">Architecture diagram:</b> The diagram displays representative nodes for readability. The labels show the actual number of features or neurons in each layer.</p>
</div>


```python
fig, ax = plt.subplots(figsize=(11, 6))

layers = [
    (0, 6, "Input Layer\n21 Features", "#2563EB"),
    (2, 6, "Hidden Layer 1\n16 Neurons", "#7C3AED"),
    (4, 5, "Hidden Layer 2\n8 Neurons", "#EC4899"),
    (6, 1, "Output Layer\n1 Prediction", "#059669")
]

positions = []

for x, visible_nodes, label, color in layers:
    if visible_nodes == 1:
        y = np.array([3])
    else:
        y = np.linspace(1, 5, visible_nodes)

    positions.append((x, y, color))

    ax.scatter(
        [x] * len(y),
        y,
        s=320,
        color=color,
        edgecolors="white",
        linewidths=2,
        zorder=3
    )

    ax.text(
        x,
        5.6,
        label,
        ha="center",
        fontsize=11,
        fontweight="bold",
        color=color
    )

for i in range(len(positions) - 1):
    x1, y1, _ = positions[i]
    x2, y2, _ = positions[i + 1]

    for start in y1:
        for end in y2:
            ax.plot(
                [x1, x2],
                [start, end],
                color="#CBD5E1",
                alpha=0.45,
                linewidth=0.8
            )

ax.text(2, 0.4, "ReLU", ha="center", color="#7C3AED")
ax.text(4, 0.4, "ReLU", ha="center", color="#EC4899")
ax.text(6, 0.4, "Sigmoid", ha="center", color="#059669")

ax.set_xlim(-0.8, 6.8)
ax.set_ylim(0, 6.5)
ax.axis("off")

plt.title("Proposed Heart Disease Neural Network", fontsize=15, pad=18)
plt.show()
```

  <h2 style="color:#FFFFFF;margin:0 0 7px 0;">Computing One Forward Pass</h2>
  <p style="color:#F5F3FF;line-height:1.7;margin:0;">I will use a small example network to follow the values from the inputs through a hidden layer and into the final prediction.</p>
</div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;border-radius:10px;padding:12px 15px;margin:12px 0;">
  <p style="color:#374151;line-height:1.8;margin:0;"><b style="color:#1D4ED8;">Example architecture:</b> 2 inputs → 2 hidden neurons using ReLU → 1 output neuron using Sigmoid.</p>
</div>


```python
inputs = np.array([0.8, -0.4])

hidden_weights = np.array([
    [0.6, -0.4],
    [-0.5, 0.7]
])

hidden_bias = np.array([0.1, 0.05])

output_weights = np.array([0.9, 0.3])
output_bias = -0.2
```

```python
hidden_sum = np.dot(inputs, hidden_weights) + hidden_bias
hidden_output = relu(hidden_sum)

output_sum = np.dot(hidden_output, output_weights) + output_bias
prediction = sigmoid(output_sum)

print("Hidden layer weighted sums:", hidden_sum)
print("Hidden layer after ReLU:", hidden_output)
print("Output weighted sum:", round(output_sum, 3))
print("Final prediction:", round(prediction, 3))
```

<div style="background:#F0FDF4;border:1px solid #BBF7D0;border-left:5px solid #22C55E;border-radius:11px;padding:14px 17px;margin:14px 0;">
  <h3 style="color:#15803D;margin:0 0 7px 0;">What Happened During the Forward Pass?</h3>
  <p style="color:#374151;line-height:1.8;margin:0;">The hidden layer produced <b style="color:#7C3AED;">0.78</b> and <b style="color:#DB2777;">-0.55</b>. ReLU kept the positive value and changed the negative value to zero. The output neuron produced <b style="color:#2563EB;">0.502</b>, and Sigmoid converted it into a probability of <b style="color:#15803D;">0.623</b>.</p>
  <p style="color:#6B7280;line-height:1.7;margin:8px 0 0 0;">This is a simplified numerical example, not a prediction from a trained patient model.</p>
</div>

  <p style="color:#FCE7F3;font-weight:bold;letter-spacing:1px;margin:0 0 6px 0;">HANDS-ON LAB · STEP 4</p>
  <h2 style="color:#FFFFFF;margin:0 0 7px 0;">Documenting the Forward Pass and Prediction Error</h2>
  <p style="color:#FFF7ED;line-height:1.7;margin:0;">I will connect the predicted probability to the true class, calculate Binary Cross-Entropy, and document the final project choices.</p>
</div>


```python
actual_label = 1

loss = -(
    actual_label * np.log(prediction)
    + (1 - actual_label) * np.log(1 - prediction)
)

print("Actual class:", actual_label)
print("Predicted probability:", round(prediction, 3))
print("Binary cross-entropy loss:", round(loss, 3))
```
