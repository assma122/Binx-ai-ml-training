1. Why Activation Functions Matter
</h2>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0;
">

A neuron first calculates a
<b style="color:#7C3AED;">weighted sum</b>.

An activation function then transforms that value before it moves to the next layer.

</p>

</div>

<div style="
    background:#F8FAFC;
    border:1px solid #E2E8F0;

<h3 style="
    color:#334155;
    margin:0 0 12px 0;
">
A Small Example
</h3>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0 0 6px 0;
">

<b style="color:#7C3AED;">First layer:</b>
<code>h = 2x + 3</code>

</p>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0 0 6px 0;
">

<b style="color:#DB2777;">Second layer:</b>
<code>output = 4h - 1</code>

</p>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0;
">

<b style="color:#059669;">After substitution:</b>
<code>output = 8x + 11</code>

</p>

</div>

<div style="

">

<p style="
    color:#7F1D1D;
    line-height:1.8;
    margin:0;
">

<b>Key idea:</b>
Two linear layers without an activation behave like one linear layer.

</p>

">
2. Common Activation Functions
</h2>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0;
">

Different activation functions transform neuron outputs in different ways.

The correct choice depends on the
<b style="color:#DB2777;">layer</b>
and the
<b style="color:#7C3AED;">prediction task</b>.

</p>

</div>

<div style="
    background:#F5F3FF;

<h3 style="
    color:#6D28D9;
    margin:0 0 10px 0;
">
ReLU — Hidden Layer Activation
</h3>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0 0 8px 0;
">

ReLU returns zero for negative inputs and keeps positive inputs unchanged.

</p>

<p style="
    color:#4C1D95;
    font-weight:bold;
    margin:0 0 8px 0;
">

ReLU(z) = max(0, z)

</p>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0;
">

<b style="color:#7C3AED;">Examples:</b>
<code>ReLU(-3) = 0</code>
and
<code>ReLU(5) = 5</code>

</p>

</div>

<div style="
    background:#EFF6FF;
    border:1px solid #BFDBFE;
    border-radius:14px;
    padding:17px 20px;
    margin:15px 0;
">

<h3 style="
    color:#1D4ED8;
    margin:0 0 10px 0;
">
Sigmoid — Binary Classification Output
</h3>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0 0 8px 0;
">

Sigmoid converts a neuron output into a value between
<b style="color:#2563EB;">0</b>
and
<b style="color:#2563EB;">1</b>.

</p>

<p style="
    color:#1E40AF;
    font-weight:bold;
    margin:0 0 8px 0;
">

sigmoid(z) = 1 / (1 + e⁻ᶻ)

</p>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0;
">

For my project, an output of
<b style="color:#DB2777;">0.81</b>
represents an estimated
<b style="color:#DB2777;">81% probability</b>
of heart disease.

</p>

</div>

<div style="
    background:#ECFDF5;
    border:1px solid #A7F3D0;
    border-radius:14px;
    padding:17px 20px;
    margin:15px 0;
">

<h3 style="
    color:#047857;
    margin:0 0 10px 0;
">
Tanh — Zero-Centered Activation
</h3>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0 0 8px 0;
">

Tanh transforms inputs into values between
<b style="color:#059669;">-1</b>
and
<b style="color:#059669;">1</b>.

</p>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0;
">

Unlike sigmoid, its output can be both negative and positive, which makes it centered around zero.

</p>

</div>

<div style="
    background:#FFF7ED;
    border:1px solid #FED7AA;
    border-radius:14px;
    padding:17px 20px;
    margin:15px 0;
">

<h3 style="
    color:#C2410C;
    margin:0 0 10px 0;
">
Softmax — Multi-Class Classification Output
</h3>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0 0 8px 0;
">

Softmax converts several output values into class probabilities that add up to
<b style="color:#EA580C;">1</b>.

</p>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0;
">

It is useful for
<b style="color:#C2410C;">multi-class classification</b>,
but my heart disease project uses
<b style="color:#DB2777;">binary classification</b>,
so sigmoid is the simpler output choice.

</p>

</div>

3. Forward Propagation
</h2>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0;

Forward propagation is the process of moving patient features through the network until the output layer produces a prediction.

</div>

How Does a Prediction Move Through the Network?
</h3>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0 0 8px 0;
">

<b style="color:#7C3AED;">Input layer:</b>
Receives the 21 processed patient features.

</p>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0 0 8px 0;

<b style="color:#DB2777;">First hidden layer:</b>
Applies weights, bias, and ReLU to produce 16 outputs.

    margin:0 0 8px 0;

<b style="color:#2563EB;">Second hidden layer:</b>
Processes the previous outputs using 8 neurons and ReLU.

<p style="

<b style="color:#059669;">Output layer:</b>

4. Understanding the Loss Function
</h2>

<p style="
    color:#374151;
    line-height:1.8;
    margin:0;
">

A loss function compares the predicted output with the true label and measures how wrong the prediction is.

</p>

</div>

<div style="
    background:#F8FAFC;
    border:1px solid #E2E8F0;
    border-radius:14px;

<h3 style="
    color:#334155;
    margin:0 0 12px 0;
">
Prediction vs. Loss
</h3>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0 0 8px 0;
">

<b style="color:#2563EB;">Prediction:</b>

The probability produced by the neural network.

</p>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0 0 8px 0;
">

<b style="color:#DB2777;">True label:</b>

The actual outcome stored in the dataset.

</p>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0;
">

<b style="color:#DC2626;">Loss:</b>

A numerical value describing the difference between the prediction and the actual outcome.

</p>

</div>

<div style="
    background:#FAF5FF;
    border:1px solid #E9D5FF;
    border-radius:14px;

<h3 style="
    color:#6D28D9;
    margin:0 0 12px 0;
">
Choosing the Right Loss Function
</h3>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0 0 9px 0;
">

<b style="color:#7C3AED;">Regression:</b>

Mean Squared Error measures the squared difference between predicted and actual numerical values.

</p>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0 0 9px 0;
">

<b style="color:#DB2777;">Binary classification:</b>

Binary Cross-Entropy evaluates predictions with two possible classes.

</p>

<p style="
    color:#374151;
    line-height:1.9;
    margin:0;
">

<b style="color:#2563EB;">Multi-class classification:</b>

Categorical Cross-Entropy evaluates probabilities across several classes.

</p>

</div>

<div style="
    background:#ECFDF5;
    border:1px solid #A7F3D0;

">

<b style="color:#7C3AED;">y:</b>

</p>

<p style="
    color:#374151;

    margin:0;
">

<b style="color:#DB2777;">p:</b>

The predicted probability of heart disease.

```python
x_values = np.linspace(-6, 6, 300)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].plot(
    x_values,
    relu(x_values),
    color="#7C3AED",
    linewidth=3
)
axes[0].set_title("ReLU")
axes[0].set_xlabel("Input")
axes[0].set_ylabel("Output")
axes[0].grid(alpha=0.2)

axes[1].plot(
    x_values,
    sigmoid(x_values),
    color="#F43F5E",
    linewidth=3
)
axes[1].set_title("Sigmoid")
axes[1].set_xlabel("Input")
axes[1].set_ylabel("Output")
axes[1].grid(alpha=0.2)

axes[2].plot(
    x_values,
    tanh(x_values),
    color="#059669",
    linewidth=3
)
axes[2].set_title("Tanh")
axes[2].set_xlabel("Input")
axes[2].set_ylabel("Output")
axes[2].grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

<h2 style="color:#6D28D9;margin:0 0 10px 0;">
Why Not Use Sigmoid in Every Layer?
</h2>

<p style="color:#374151;line-height:1.8;margin:0;">
Sigmoid is useful for the output layer because it produces a probability between 0 and 1. However, using it repeatedly inside hidden layers can make learning slower.
</p>

<p style="color:#374151;line-height:1.8;margin:10px 0 0 0;">
When Sigmoid outputs become very close to 0 or 1, the learning signal can become very small. Across multiple layers, this may lead to the
<b style="color:#BE185D;">vanishing gradient problem</b>.
</p>

<p style="color:#374151;line-height:1.8;margin:10px 0 0 0;">
<b style="color:#059669;">ReLU</b>
is commonly preferred in hidden layers because it is simple and helps preserve stronger learning signals for positive inputs.
</p>

</div>

<div style="
    background:#FFF7ED;
    border-left:5px solid #F97316;
    border-radius:11px;
    padding:14px 17px;
    margin:14px 0;
">

<p style="color:#374151;line-height:1.8;margin:0;">
<b style="color:#C2410C;">A limitation of ReLU:</b>
A neuron that repeatedly receives negative inputs can stay inactive because ReLU returns zero for those values. This is sometimes called the
<b style="color:#BE185D;">dying ReLU problem</b>.
</p>

</div>

<div style="
    background:#FFFFFF;
    border:1px solid #E5E7EB;

    margin:19px 0;
">

<h2 style="color:#4338CA;margin:0 0 13px 0;">

<h2 style="color:#C2410C;margin:0 0 10px 0;">
Why Not Use Mean Squared Error?
</h2>

<p style="color:#374151;line-height:1.8;margin:0;">
Mean Squared Error is commonly used when the model predicts a continuous value, such as a house price.
</p>

<p style="color:#374151;line-height:1.8;margin:10px 0 0 0;">
In this project, the model predicts the probability of one of two classes.
<b style="color:#BE185D;">Binary Cross-Entropy</b>
is more suitable because it evaluates probability-based binary predictions and strongly penalizes confident incorrect results.
</p>

</div>

<div style="
    background:#FDF2F8;
    border-left:5px solid #EC4899;
    border-radius:11px;
    padding:14px 17px;

">

<p style="color:#374151;line-height:1.8;margin:0;">
<b style="color:#BE185D;">Medical example:</b>
If a patient actually has heart disease, predicting a very low disease probability should produce a large loss because the model is confidently wrong.
</p>

</div>

<h2 style="color:white;margin:0 0 8px 0;">
Visualizing Activation Functions
</h2>

<p style="color:#F5F3FF;line-height:1.7;margin:0;">
I will plot <b>ReLU</b>, <b>Sigmoid</b>, and <b>Tanh</b> to observe how each function transforms the same input values.
</p>

</div>

```python
import numpy as np
import matplotlib.pyplot as plt
def relu(x):
    return np.maximum(0, x)


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def tanh(x):
    return np.tanh(x)


x_values = np.linspace(-6, 6, 300)
```

The output layer will contain one neuron using Sigmoid. It converts the final weighted sum into a probability between 0 and 1.

<h2 style="color:white;margin:0 0 8px 0;">
Computing One Forward Pass
</h2>

<p style="color:#F5F3FF;line-height:1.7;margin:0;">
I will use a small example network to follow the values from the inputs through the hidden layer and into the final prediction.
</p>

</div>

<div style="
    background:#EFF6FF;
    border-left:5px solid #3B82F6;
    border-radius:11px;
    padding:14px 17px;
    margin:15px 0;
">

<p style="color:#374151;line-height:1.8;margin:0;">
<b style="color:#1D4ED8;">Example architecture:</b>
2 inputs → 2 hidden neurons using ReLU → 1 output neuron using Sigmoid.
</p>

</div>

```python
inputs = np.array([0.8, -0.4])

hidden_weights = np.array([
    [0.6, -0.4],
    [-0.5, 0.7]
])

hidden_bias = np.array([0.1, 0.05])

output_weights = np.array([0.9, 0.3])

output_bias = -0.2
```

```python
hidden_sum = np.dot(inputs, hidden_weights) + hidden_bias

hidden_output = relu(hidden_sum)

output_sum = np.dot(hidden_output, output_weights) + output_bias

prediction = sigmoid(output_sum)

print("Hidden layer weighted sums:", hidden_sum)
print("Hidden layer after ReLU:", hidden_output)
print("Output weighted sum:", round(output_sum, 3))
print("Final prediction:", round(prediction, 3))
```

<h3 style="color:#15803D;margin:0 0 8px 0;">
What Happened During the Forward Pass?

<p style="color:#374151;line-height:1.8;margin:0;">
The hidden layer produced two values:
<b style="color:#7C3AED;">0.78</b> and

ReLU kept the positive value and changed the negative value to zero.
The output neuron then combined the hidden results, and Sigmoid converted the final value into a probability of
<b style="color:#15803D;">0.623</b>.

<h2 style="color:white;margin:0 0 8px 0;">
Measuring the Prediction Error
</h2>

<p style="color:#FFF7ED;line-height:1.7;margin:0;">
After the forward pass produces a prediction, Binary Cross-Entropy measures how far that prediction is from the actual class.
</p>

</div>

```python
actual_label = 1

loss = -(
    actual_label * np.log(prediction)
    + (1 - actual_label) * np.log(1 - prediction)
)

print("Actual class:", actual_label)
print("Predicted probability:", round(prediction, 3))
print("Binary cross-entropy loss:", round(loss, 3))
```

<h2 style="color:#1D4ED8;margin:0 0 7px 0;">
How Does Confidence Affect the Loss?

<p style="color:#374151;line-height:1.7;margin:0;">
I will compare three predicted probabilities for a patient whose actual class is 1.

```python
for probability in [0.10, 0.50, 0.90]:
    loss_value = -np.log(probability)

    print(
        f"Prediction: {probability:.2f} "
        f"| Loss: {loss_value:.3f}"
    )
```
