## Why Do Neural Networks Need Activation Functions?

A neuron first combines its inputs using **weights** and a **bias**:

<div style="
    background:#F6F3FF;
    border-radius:9px;
    padding:12px;
    margin:12px 0;
    text-align:center;
    color:#4C1D95;
    font-size:18px;
">

**z = XW + b**

</div>

If every layer only performs linear calculations like this, adding more layers does not make the network much more powerful.

A sequence of linear operations can still behave like one linear transformation.

This means the network would struggle to learn more complex relationships in the data.

<div style="
    background:#FFF1F2;
    border-left:4px solid #F43F5E;
    border-radius:8px;
    padding:12px 14px;
    margin:14px 0;
">

<b style="color:#BE123C;">The key idea:</b><br>
Activation functions introduce <b>non-linearity</b>, allowing neural networks to learn complex patterns instead of only simple linear relationships.

</div>

## Linear vs Non-Linear Network

<div style="
    background:#FAFAFF;
    border-radius:10px;
    padding:15px;
    margin:12px 0;
">

<b style="color:#F43F5E;">Without Activation Functions</b>

Input → Linear Layer → Linear Layer → Output

<br>

<b style="color:#7C3AED;">With Activation Functions</b>

Input → Linear Layer → Activation → Linear Layer → Activation → Output

</div>

For clinical data, relationships between features such as **age, cholesterol, heart rate, chest pain, and exercise measurements** may be complex.

Activation functions allow the network to model these relationships more flexibly.

## What Does an Activation Function Actually Do?

After a neuron calculates its weighted sum:

<div style="
    background:#F6F3FF;
    border-radius:9px;
    padding:11px;
    margin:10px 0;
    text-align:center;
    color:#4C1D95;
">

**z = XW + b**

</div>

the result is passed through an activation function:

<div style="
    background:#FFF1F2;
    border-radius:9px;
    padding:11px;
    margin:10px 0;
    text-align:center;
    color:#9F1239;
">

**a = Activation(z)**

</div>

Here:

- **z** is the weighted sum produced by the neuron.
- **a** is the activated output.
- This output is passed forward to the next layer.

The activation function controls how each neuron responds to the information it receives.

## Common Activation Functions

Neural networks use different activation functions depending on the layer and the task.

The four most common functions in this lesson are:

<div style="
    background:#FAFAFF;
    border-radius:10px;
    padding:14px 16px;
    margin:12px 0;
">

<span style="color:#7C3AED;"><b>ReLU</b></span> — commonly used in hidden layers.  

<span style="color:#F43F5E;"><b>Sigmoid</b></span> — commonly used for binary classification outputs.  

<span style="color:#7C3AED;"><b>Tanh</b></span> — produces values centered around zero.  

<span style="color:#F43F5E;"><b>Softmax</b></span> — commonly used for multi-class classification outputs.

</div>

## ReLU Activation Function

**ReLU** stands for **Rectified Linear Unit**.

It is one of the most widely used activation functions in hidden layers.

<div style="
    background:#F6F3FF;
    border-radius:9px;
    padding:12px;
    margin:12px 0;
    text-align:center;
    color:#4C1D95;
    font-size:17px;
">

**ReLU(x) = max(0, x)**

</div>

This means:

- Negative inputs become **0**.
- Positive inputs remain unchanged.

<div style="
    background:#FAFAFF;
    border-radius:9px;
    padding:12px 14px;
    margin:12px 0;
">

**Example**

- ReLU(-4) = 0
- ReLU(-1) = 0
- ReLU(0) = 0
- ReLU(3) = 3
- ReLU(7) = 7

</div>

ReLU is commonly used in **hidden layers** because it is simple, computationally efficient, and works well in many neural networks.

<div style="
    background:#FFF1F2;
    border-left:4px solid #F43F5E;
    border-radius:8px;
    padding:12px 14px;
    margin:14px 0;
">

<b style="color:#BE123C;">Key idea:</b><br>
ReLU removes negative values while keeping positive values active, which allows hidden neurons to pass useful signals forward through the network.

</div>

## Sigmoid Activation Function

The **Sigmoid** function transforms any input into a value between **0 and 1**.

<div style="
    background:#F6F3FF;
    border-radius:9px;
    padding:12px;
    margin:12px 0;
    text-align:center;
    color:#4C1D95;
    font-size:17px;
">

**Sigmoid(x) = 1 / (1 + e⁻ˣ)**

</div>

Because its output is between 0 and 1, Sigmoid is useful when the neural network output represents a **probability**.

For example, an output of:

<div style="
    background:#FAFAFF;
    border-radius:9px;
    padding:12px;
    margin:12px 0;
    text-align:center;
">

**0.82**

</div>

can be interpreted as a high predicted probability for the positive class.

For this reason, Sigmoid is commonly used in the **output layer of binary classification models**.

<div style="
    background:#FFF1F2;
    border-left:4px solid #F43F5E;
    border-radius:8px;
    padding:12px 14px;
    margin:14px 0;
">

<b style="color:#BE123C;">Connection to the Heart Disease Project:</b><br>
The target has two classes: <b>HeartDisease = 0</b> or <b>HeartDisease = 1</b>.

Therefore, a Sigmoid activation is a natural choice for the final output neuron because it produces a value between 0 and 1.

</div>

## Tanh Activation Function

The **Tanh** activation function transforms inputs into values between **-1 and 1**.

<div style="
    background:#F6F3FF;
    border-radius:9px;
    padding:12px;
    margin:12px 0;
    text-align:center;
    color:#4C1D95;
    font-size:17px;
">

**Tanh Output Range: -1 to +1**

</div>

Unlike Sigmoid, Tanh is centered around zero.

This can sometimes help when a hidden layer benefits from having both negative and positive activation values.

Tanh can be used in hidden layers, although **ReLU is usually the more common default choice**.

## Softmax Activation Function

**Softmax** is mainly used in the output layer of **multi-class classification models**.

It converts the output scores into probabilities.

Each probability is between **0 and 1**, and all probabilities add up to **1**.

<div style="
    background:#FAFAFF;
    border-radius:9px;
    padding:12px 14px;
    margin:12px 0;
">

**Example**

Class A → 0.10  
Class B → 0.75  
Class C → 0.15  

**Total = 1.00**

</div>

The class with the highest probability is usually selected as the final prediction.

## Visualizing the Activation Functions

Reading the equations explains what each activation function does, but plotting them makes the differences much easier to see.

In this section, I will compare **ReLU**, **Sigmoid**, and **Tanh** over the same input range.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)
```

```python
x = np.linspace(-10, 10, 400)
```

```python
plt.figure(figsize=(9, 5))

plt.plot(x, relu(x), label="ReLU", linewidth=2.5)
plt.plot(x, sigmoid(x), label="Sigmoid", linewidth=2.5)
plt.plot(x, tanh(x), label="Tanh", linewidth=2.5)

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.title("Activation Functions")
plt.xlabel("Input Value")
plt.ylabel("Activated Output")

plt.legend()
plt.grid(alpha=0.25)

plt.show()
```

## Forward Propagation

After the activation functions are defined, the next question is:

<div style="

">

<b style="color:#BE123C;">How does the network actually produce a prediction?</b>

</div>

The answer is **forward propagation**.

Forward propagation is the process of moving the input data through the neural network, layer by layer, until the final prediction is produced.

## How Does Information Move Through the Network?

At each layer, the network performs the same basic sequence:

<div style="
    background:#FAFAFF;
    border-radius:10px;
    padding:14px 16px;
    margin:12px 0;
">

<span style="color:#7C3AED;"><b>1. Receive inputs</b></span><br>
The layer receives values from the previous layer.

<span style="color:#F43F5E;"><b>2. Apply weights and bias</b></span><br>
The inputs are combined using learned weights and a bias.

<span style="color:#7C3AED;"><b>3. Apply an activation function</b></span><br>
The weighted sum is transformed using an activation such as ReLU or Sigmoid.

<span style="color:#F43F5E;"><b>4. Pass the result forward</b></span><br>
The activated output becomes the input to the next layer.

</div>

## Forward Pass Flow

<div style="
    background:#F6F3FF;

">

**Input Features**  
↓  
**Weighted Sum + Bias**  
↓  
**ReLU Activation**  
↓  
**Hidden Layer**  
↓  
**Weighted Sum + Bias**  
↓  
**Sigmoid Output**  
↓  
**Prediction**

</div>

## Forward Propagation Mathematically

The same process can be written as:

<div style="
    background:#F6F3FF;
    border-radius:9px;
    padding:14px;
    margin:12px 0;
    color:#4C1D95;
">

**Layer 1:**  
z₁ = XW₁ + b₁  
a₁ = ReLU(z₁)

**Layer 2:**  
z₂ = a₁W₂ + b₂  
a₂ = ReLU(z₂)

**Output Layer:**  
z₃ = a₂W₃ + b₃  
ŷ = Sigmoid(z₃)

<b style="color:#BE123C;">Key Idea:</b><br>

Forward propagation is simply the network moving information from the input layer to the output layer to produce a prediction.

No learning happens during this step alone — it is only the calculation of the current prediction.

</div>

## A Tiny Forward Pass Example

To make forward propagation more concrete, I will compute one small forward pass using NumPy.

The network will contain:

<div style="
    background:#F6F3FF;
    border:1px solid #DDD6FE;
    border-radius:10px;
    padding:14px;
    margin:12px 0;
    text-align:center;
    color:#4C1D95;
">

**2 Inputs → 2 Hidden Neurons → 1 Output Neuron**

</div>

The hidden layer will use **ReLU**, while the output neuron will use **Sigmoid**.

```python
# Sample input
X = np.array([0.6, 0.9])

# Hidden layer weights and bias
W1 = np.array([
    [0.5, -0.4],
    [0.8,  0.3]
])

b1 = np.array([0.1, -0.2])

# Output layer weights and bias
W2 = np.array([0.7, -0.5])
b2 = 0.2
```

```python
# Hidden layer
z1 = np.dot(X, W1) + b1
a1 = relu(z1)

print("Hidden weighted sum:", z1)
print("Hidden activation:", a1)
```

```python
# Output layer
z2 = np.dot(a1, W2) + b2
prediction = sigmoid(z2)

print("Output weighted sum:", round(z2, 3))
print("Prediction:", round(prediction, 3))
```

<div style="
    background:#FFF1F2;
    border-left:4px solid #F43F5E;
    border-radius:8px;
    padding:13px 15px;
    margin:14px 0;
">

<b style="color:#BE123C;">How to Read the Forward Pass:</b><br><br>

The input values are first combined using the hidden-layer weights and bias.

The result is passed through <b style="color:#7C3AED;">ReLU</b>.

The activated hidden values then move to the output neuron.

Finally, the output neuron applies <b style="color:#F43F5E;">Sigmoid</b> to produce a value between 0 and 1.

</div>

### Observation

The forward pass transformed the original input values through a hidden layer and then produced a final probability-like output.

This shows how a neural network prediction is built progressively from **inputs → weighted sums → activations → output**.

## Loss Functions

After the forward pass produces a prediction, the network needs a way to measure how far that prediction is from the correct answer.

This is the role of the **loss function**.

<div style="
    background:#FFF1F2;
    border-left:4px solid #F43F5E;
    border-radius:8px;
    padding:12px 14px;
    margin:14px 0;
">

<b style="color:#BE123C;">Main Idea:</b><br>
A <b>small loss</b> means the prediction is close to the true value, while a <b>large loss</b> means the prediction is far from the correct answer.

</div>

During training, the network tries to reduce the loss as much as possible.

## Mean Squared Error

**Mean Squared Error (MSE)** is commonly used for regression problems.

It measures the average squared difference between the predicted value and the true value.

<div style="
    background:#F6F3FF;
    border-radius:9px;
    padding:12px;
    margin:12px 0;
    text-align:center;
    color:#4C1D95;
">

**MSE = average of (true value − predicted value)²**

</div>

Because the differences are squared, larger prediction errors receive a stronger penalty.

## Binary Cross-Entropy

**Binary Cross-Entropy** is used when the target contains two classes.

For example:

<div style="
    background:#FAFAFF;
    border-radius:9px;
    padding:12px 14px;
    margin:12px 0;
">

**0 → No Heart Disease**  
**1 → Heart Disease**

</div>

It compares the predicted probability with the true binary label.

Binary cross-entropy works naturally with a **Sigmoid output**, because Sigmoid produces a value between 0 and 1.

<div style="
    background:#FFF1F2;
    border-left:4px solid #F43F5E;
    border-radius:8px;

<b style="color:#BE123C;">Project Choice:</b><br>

The heart disease project is a <b>binary classification</b> problem.

Therefore, the most suitable combination is:

<br><br>

<b style="color:#7C3AED;">Output Activation:</b> Sigmoid<br>
<b style="color:#F43F5E;">Loss Function:</b> Binary Cross-Entropy

## Categorical Cross-Entropy

**Categorical Cross-Entropy** is used for multi-class classification problems.

For example, if a model predicts one class from:

<div style="

the output layer can use **Softmax** to produce a probability for each class.

Categorical cross-entropy then measures how well those predicted probabilities match the correct class.
