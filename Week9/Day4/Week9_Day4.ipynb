<div style="font-size:11px;font-weight:800;letter-spacing:1.6px;color:#059669;margin-bottom:7px;">4.3 · MODEL & PREPROCESSING FILES</div>
<div style="font-size:23px;font-weight:900;color:#1F2937;margin-bottom:10px;">Keep Every Runtime <span style="color:#10B981;">Artifact Together</span></div>
<div style="font-size:14px;line-height:1.8;color:#4B5563;">The model and vectorizer are stored beside <code>app.py</code>, so all paths are relative and work locally and on the server. The app also reuses the exact NLTK preprocessing from training before applying TF-IDF, preventing training-serving skew.</div>
</div>

```python
from pathlib import Path

required_files = [
    "app.py",
    "model.joblib",
    "vectorizer.joblib",
    "requirements.txt",
    "README.md",
]

for filename in required_files:
    path = Path(filename)
    status = "READY" if path.exists() and path.stat().st_size > 0 else "MISSING"
    print(f"{filename:<22} {status:>7}")
```

```python
import joblib

model = joblib.load("model.joblib")
vectorizer = joblib.load("vectorizer.joblib")

print("Model:", type(model).__name__)
print("Vectorizer:", type(vectorizer).__name__)
print("Vocabulary size:", len(vectorizer.vocabulary_))
print("Classes:", list(model.classes_))
```

<div style="font-size:11px;font-weight:800;letter-spacing:1.6px;color:#D97706;margin-bottom:7px;">LOCAL VALIDATION</div>
<div style="font-size:23px;font-weight:900;color:#1F2937;margin-bottom:10px;">Test the App Before <span style="color:#F59E0B;">Going Live</span></div>
<div style="font-size:14px;line-height:1.8;color:#4B5563;">Streamlit's testing interface runs the real application, interacts with its widgets, and verifies that the positive, negative, and empty-input paths complete without exceptions.</div>
</div>

```python
from streamlit.testing.v1 import AppTest

app_test = AppTest.from_file("app.py", default_timeout=60)
app_test.run()

print("Initial load:", "PASS" if not app_test.exception else "FAIL")

app_test.selectbox[0].set_value("Positive example").run()
app_test.button[0].click().run()
print("Positive example:", app_test.success[0].value)

app_test.selectbox[0].set_value("Negative example").run()
app_test.button[0].click().run()
print("Negative example:", app_test.error[0].value)

app_test.selectbox[0].set_value("Write my own review").run()
app_test.text_area[0].set_value("").run()
app_test.button[0].click().run()
print("Empty input:", app_test.warning[0].value)
print("Exceptions:", len(app_test.exception))
```
