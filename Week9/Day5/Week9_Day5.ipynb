<div style="font-family:Georgia,serif;font-size:30px;font-weight:700;color:#4A2030;margin:0 0 18px 0;padding-bottom:10px;border-bottom:2px solid #D9B875;">5.2 Technical write-up</div>

<div style="font-size:18px;font-weight:900;color:#7A2D48;margin:20px 0 10px 0;">Problem</div>

Movie platforms receive more written feedback than a person can inspect manually. CineSense provides a first-pass classification of an English movie review into positive or negative sentiment and displays the model's confidence.

<div style="font-size:18px;font-weight:900;color:#7A2D48;margin:20px 0 10px 0;">Dataset and split</div>

| Item | Value |
|---|---:|
| Dataset | IMDb movie reviews |
| Records after duplicate removal | **49,582** |
| Training records | **39,665** |
| Test records | **9,917** |
| Split | Stratified 80/20 |
| Random seed | `42` |

<div style="font-size:18px;font-weight:900;color:#7A2D48;margin:20px 0 10px 0;">Method</div>

1. Decode HTML and remove tags.
2. Normalize negative contractions and lowercase the text.
3. Remove digits and punctuation.
4. Tokenize and remove stop words while preserving `not`, `no`, and `nor`.
5. Apply POS-aware WordNet lemmatization.
6. Convert text to TF-IDF vectors with at most 5,000 features.
7. Train Logistic Regression with `max_iter=1000` and `random_state=42`.

</div>


<div style="padding:26px;border-radius:20px;background:linear-gradient(135deg,#351D26,#6F263D);color:#FFF9F2;margin:28px 0;box-shadow:0 16px 34px rgba(71,26,43,.18);">
<div style="font-size:11px;font-weight:900;letter-spacing:2px;color:#E8C98E;margin-bottom:14px;">HELD-OUT TEST PERFORMANCE</div>
<div style="display:flex;gap:12px;flex-wrap:wrap;">
<div style="flex:1;min-width:130px;padding:16px;border-radius:14px;background:rgba(255,255,255,.09);"><div style="font-size:12px;color:#E7D8D8;">Accuracy</div><div style="font-size:30px;font-weight:900;color:#FFFFFF;">88.62%</div></div>
<div style="flex:1;min-width:130px;padding:16px;border-radius:14px;background:rgba(255,255,255,.09);"><div style="font-size:12px;color:#E7D8D8;">Precision</div><div style="font-size:30px;font-weight:900;color:#FFFFFF;">87.76%</div></div>
<div style="flex:1;min-width:130px;padding:16px;border-radius:14px;background:rgba(255,255,255,.09);"><div style="font-size:12px;color:#E7D8D8;">Recall</div><div style="font-size:30px;font-weight:900;color:#FFFFFF;">89.85%</div></div>

</div>
